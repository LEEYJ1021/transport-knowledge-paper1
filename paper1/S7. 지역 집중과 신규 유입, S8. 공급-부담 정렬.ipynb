````python
# -*- coding: utf-8 -*-
# =====================================================================================
# S7 지역 집중과 신규 유입 (단일 셀, OpenAlex 호출 없음). 미실행 코드: QUICK=True로 먼저 시험.
# 사용: MODE="freeze"(규칙 동결, 데이터 미열람) -> MODE="run"(동결 해시 검증 후 분석)
# 입력: 02_S2/country_mass_long, 01_S1(wb_income_current/history, authorship_level, authorship_level_recovered),
#       analysis_data(F1, F2_risk_names, paper_author_long), 03_S3/L7_topic_grade, 00_S0/PREREG
# 산출: paper1/07_S7/*.csv, S7_meta.json. 해석은 '함께 움직인다'까지, '체제' 대신 '기간'.
# =====================================================================================
import json, hashlib, subprocess, warnings
from pathlib import Path
import numpy as np, pandas as pd, scipy.sparse as sp
from scipy.stats import spearmanr
from scipy.spatial.distance import jensenshannon
warnings.filterwarnings("ignore"); pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40); pd.set_option("display.max_rows", 200)

MODE = "run"      # "freeze" | "run"
QUICK = False         # True: 부트스트랩 B=50 (시험용, 파일명에 _QUICK)
COMMIT = True       # run 끝에서 True여야 git 커밋·태그

ROOT = Path.home() / "Research/Minnesota_Transport"
DATA, S0, S1, S2, S3, S7 = ROOT/"analysis_data/final", ROOT/"paper1/00_S0", ROOT/"paper1/01_S1", ROOT/"paper1/02_S2", ROOT/"paper1/03_S3", ROOT/"paper1/07_S7"
PAL = ROOT/"analysis_data/paper_author_long.csv"
now = lambda: pd.Timestamp.now().isoformat(timespec="seconds")
sha_b = lambda b: hashlib.sha256(b).hexdigest(); sha_f = lambda p: sha_b(Path(p).read_bytes())
tf = lambda s: s.astype(str).str.strip().str.lower().isin(["1", "1.0", "true", "t", "yes"])
class Halt(Exception): pass
def git(*a): return subprocess.run(["git", "-C", str(ROOT), *a], capture_output=True, text=True)

SPEC = dict(
    version=1, seed=2026, decision_status="모든 수치는 임의 출발값. S7 결과 열람 전에 동결. S5에서 체제 서사가 철회되어 '기간'으로만 비교.",
    periods={"P0709": [2007, 2009], "P1014": [2010, 2014], "P1519": [2015, 2019], "P2025": [2020, 2025]},
    period_note="국가 층 창이 2007년부터라 첫 기간이 3년(S4의 5개 기간과 다름). 마지막 기간은 6년(2025 부분연도·보유율 하락 후보).",
    main=dict(scheme="inv_n_authors", variant="main", sample="all", terr_merge=False, income="fixed"),
    grid=dict(scheme=["equal", "inv_n_authors", "first_last", "full_count"], variant=["main", "rec"], sample=["all", "fixed14"],
              terr_merge=[False, True], income=["fixed", "tv"]),
    terr_merge_map={"HK": "CN", "MO": "CN"}, top_k=5, active_mass_min=1.0,
    groups=["US", "CN", "HK_MO", "HIC_other", "UMC_other", "LMIC", "NOINC", "UNK"],
    group_rule="US, CN, HK+MO 별도, 나머지는 소득군(고정=wb_income_current, 시변=OGHIST 산정연도): HIC_other, UMC_other, LMIC(LMC+LIC), 소득군 없음=NOINC, 미상·구국가=UNK",
    stats=["top5_incl", "top5_known", "hhi_known", "cn_known", "us_known", "unk", "n_active", "theil_total", "theil_between", "theil_within_share"],
    theil="국가당 동일 가중 벤치마크 T=sum s ln(s N), N=그 구성의 알려진 국가 수(인구 가중 아님). 소득군 간/내 정확 분해, 항등식 점검 1e-9",
    direction_rule="주 구성과 방향이 다른 구성 비율 <= 0.25 (추세 Spearman 부호, 첫~마지막 기간 변화 부호). theil_*는 income 차원 포함, 그 외 통계는 income=fixed만 센다",
    bootstrap=dict(B=500, kind="논문 단위 Poisson(1) 가중(저널·공저 상관 미반영: 구간은 낙관적일 수 있음)"),
    author_layer=dict(id="name_norm(paper_author_long)", risk_variants=["base", "none"], base_weight="1/n_authors(F1 저자 수), 소속 국가 균등 분할",
                      roles="지속(두 기간 모두), 신규(t1에만, 첫 출판연도 >= t1 시작), 복귀(t1에만, 첫 출판연도 < t1 시작), 퇴출(t0에만)",
                      decomposition="Griliches-Regev 중점 분해를 지역군에 적용(적응, 재배분, 신규, 복귀, 퇴출)"),
    focus_topics=[88, 45, 69, 50, 0, 147, 38], top_change_n=10,
    interpretation="소속 국가는 국적이 아님. 지역 구성 변화는 함께 움직임의 기술이며 인과 아님. 복구 국가(rec)는 민감도.",
)
def spec_hash(s): return sha_b(json.dumps({k: v for k, v in s.items() if k not in ("sha256", "created", "refs")}, sort_keys=True, ensure_ascii=False).encode())

def freeze():
    S7.mkdir(parents=True, exist_ok=True); h = spec_hash(SPEC); p = S7/"S7_0_PREREG.json"
    print("S7-0 해시(예정)", h[:12])
    if p.exists():
        old = json.loads(p.read_text(encoding="utf-8")); print("이미 동결됨:", "동일" if old["sha256"] == h else "⛔ 다름(새 버전으로 분리하고 CHANGELOG에 사유)"); return
    refs = {}
    for f in [S2/"S2_PREREG_ADDENDUM.json", S1/"RECOVERY_RULES_PREREG.json", S1/"TERRITORY_RULES.md", S3/"L7_topic_grade.csv", S0/"PREREG.json"]:
        if f.exists(): refs[f.name] = sha_f(f)
    for n in ["country_mass_long.parquet", "country_mass_long.csv.gz"]:
        if (S2/n).exists(): refs[n] = sha_f(S2/n)
    p.write_text(json.dumps(dict(SPEC, created=now(), sha256=h, refs=refs), ensure_ascii=False, indent=1), encoding="utf-8")
    (S7/"CHANGELOG_S7.md").write_text(f"# CHANGELOG S7\n\n- {now()} | S7-0 동결(해시 {h[:12]}): 기간 4개, 구성 격자(가중치4 x 변형2 x 표본2 x 영토2 x 소득2), Theil(동일 가중), 부트스트랩 B=500, 저자 층 역할 정의. "
                                      f"S1~S6 산출물은 이미 열람한 상태에서 동결(S2의 C4 부호표·상위5개국 중간 결과를 본 뒤). 모든 수치는 임의 출발값 | S7 설계 | S7 전체\n", encoding="utf-8")
    print("저장: S7_0_PREREG.json | 참조 해시", list(refs))

def run():
    p = S7/"S7_0_PREREG.json"
    if not p.exists(): raise Halt("먼저 MODE='freeze'로 동결하세요")
    sp_ = json.loads(p.read_text(encoding="utf-8"))
    if spec_hash(sp_) != sp_["sha256"] or sp_["sha256"] != spec_hash(SPEC): raise Halt("동결 파일과 현재 SPEC이 다릅니다(사후 변경 금지)")
    print(f"동결 검증 통과 {sp_['sha256'][:12]} | QUICK={QUICK}")
    rng = np.random.default_rng(int(SPEC["seed"])); sfx = "_QUICK" if QUICK else ""
    PER = {k: tuple(v) for k, v in SPEC["periods"].items()}; YP = {y: k for k, (a, b) in PER.items() for y in range(a, b+1)}
    GROUPS = SPEC["groups"]; TM = SPEC["terr_merge_map"]; TOPK = SPEC["top_k"]; AMIN = SPEC["active_mass_min"]; mn = SPEC["main"]
    FIXED = set(json.loads((S0/"PREREG.json").read_text(encoding="utf-8"))["event"]["fixed_venues"]); assert len(FIXED) == 14
    cur = pd.read_csv(S1/"wb_income_current.csv", keep_default_na=False); cur = cur[(cur.income != "INX") & (cur.region != "NA")]
    INC_FIXED = dict(zip(cur.iso2, cur.income))
    H = pd.read_csv(S1/"wb_income_history.csv", keep_default_na=False); INC_TV = {(a, int(b)): c for a, b, c in zip(H.iso2, H.year, H.income)}
    inc_of = lambda c, y, m: (lambda v: v if v in ("HIC", "UMC", "LMC", "LIC") else "NOINC")(INC_FIXED.get(c) if m == "fixed" else INC_TV.get((c, y)))
    def grp_of(c, y, m="fixed"):
        if c in ("UNK", "ZZ_former"): return "UNK"
        if c in ("US", "CN"): return c
        if c in ("HK", "MO"): return "HK_MO"
        return {"HIC": "HIC_other", "UMC": "UMC_other", "LMC": "LMIC", "LIC": "LMIC"}.get(inc_of(c, y, m), "NOINC")
    def add_group(d, m="fixed"):
        u = d[["country", "year"]].drop_duplicates(); u["g"] = [grp_of(c, y, m) for c, y in zip(u.country, u.year)]; return d.merge(u, on=["country", "year"], how="left")

    # ---------- 입력 ----------
    pm = S2/"country_mass_long.parquet"
    M = pd.read_parquet(pm) if pm.exists() else pd.read_csv(S2/"country_mass_long.csv.gz", dtype={"pid": str})
    M["pid"] = M.pid.astype(str); M = M[M.year >= 2007]
    print(f"질량표 {len(M):,}행 | scheme {sorted(M.scheme.unique())} | variant {sorted(M.variant.unique())}")
    for sc in ["equal", "inv_n_authors", "first_last"]:       # 불변식: 논문당 질량 합 1
        s = M[(M.scheme == sc) & (M.variant == "main")].groupby("pid").w.sum()
        if not np.allclose(s, 1.0): raise Halt(f"불변식 실패: {sc} 논문당 질량 합 != 1")
    print("불변식 통과: 논문당 질량 합 = 1 (equal, inv_n_authors, first_last)")
    pools = {(sc, v): d for (sc, v), d in M.groupby(["scheme", "variant"])}

    def prep(sc, v, sample, terr):
        d = pools[(sc, v)]
        if sample == "fixed14": d = d[d.venue_short.isin(FIXED)]
        d = d[["year", "country", "w"]]
        if terr: d = d.assign(country=d.country.replace(TM))
        return d.groupby(["year", "country"], as_index=False).w.sum()

    def theil_parts(s, gi, ng):
        N = len(s); nz = s > 0; T = float((s[nz]*np.log(s[nz]*N)).sum()); sg = np.bincount(gi, weights=s, minlength=len(ng)); ok = sg > 0
        Tb = float((sg[ok]*np.log(sg[ok]*N/ng[ok])).sum()); Tw = 0.0
        for g in np.where(ok)[0]:
            m = (gi == g) & nz; r = s[m]/sg[g]; Tw += sg[g]*float((r*np.log(r*ng[g])).sum())
        if abs(T - Tb - Tw) > 1e-9: raise Halt("불변식 실패: Theil 분해 항등식")
        return T, Tb, Tw
    INC_IDX = {"HIC": 0, "UMC": 1, "LMC": 2, "LIC": 3, "NOINC": 4}
    def stats_of(vec, cols, y, imode):
        unk = np.isin(cols, ["UNK", "ZZ_former"]); kn = np.where(unk, 0.0, vec); kt = kn.sum(); tot = vec.sum()
        if kt <= 0: return {}
        top = np.sort(kn)[::-1][:TOPK].sum(); s = (kn/kt)[~unk]; kc = cols[~unk]
        gi = np.array([INC_IDX[inc_of(c, y, imode)] for c in kc]); ng = np.bincount(gi, minlength=5).astype(float)
        T, Tb, Tw = theil_parts(s, gi, ng)
        return dict(top5_incl=top/tot, top5_known=top/kt, hhi_known=float(((kn/kt)**2).sum()), cn_known=float(kn[cols == "CN"].sum()/kt), us_known=float(kn[cols == "US"].sum()/kt),
                    unk=float(1 - kt/tot), n_active=int((kn >= AMIN).sum()), theil_total=T, theil_between=Tb, theil_within_share=Tw/T if T > 0 else np.nan)
    def stats_table(d, imode):
        G = d.pivot_table(index="year", columns="country", values="w", aggfunc="sum", fill_value=0); cols = np.array(G.columns)
        Y = pd.DataFrame([dict(year=int(y), **stats_of(v.values, cols, int(y), imode)) for y, v in G.iterrows()])
        Pp = pd.DataFrame([dict(period=k, **stats_of(G.loc[a:b].sum(0).values, cols, (a+b)//2, imode)) for k, (a, b) in PER.items()])
        return Y, Pp, G

    # ---------- 7.1 구성 격자: 가중치 x 변형 x 표본 x 영토 x 소득 ----------
    G_ = SPEC["grid"]; Yl, Pl, SG = [], [], []; mainG = None
    for sc in G_["scheme"]:
        for v in G_["variant"]:
            for sample in G_["sample"]:
                for terr in G_["terr_merge"]:
                    d = prep(sc, v, sample, terr)
                    for im in G_["income"]:
                        Y, Pp, G = stats_table(d, im); cfg = dict(scheme=sc, variant=v, sample=sample, terr_merge=terr, income=im)
                        Yl.append(Y.assign(**cfg)); Pl.append(Pp.assign(**cfg))
                        for st in SPEC["stats"]:
                            y = Y[st].astype(float); a = Pp.set_index("period")[st]
                            SG.append(dict(**cfg, stat=st, rho=spearmanr(Y.year, y)[0], slope=float(np.polyfit(Y.year, y, 1)[0]), delta_first_last=float(a["P2025"]-a["P0709"]),
                                           sign_trend=int(np.sign(spearmanr(Y.year, y)[0])), sign_delta=int(np.sign(a["P2025"]-a["P0709"]))))
                        if (sc, v, sample, terr, im) == (mn["scheme"], mn["variant"], mn["sample"], mn["terr_merge"], mn["income"]): mainG = G
    Y_all, P_all, S_all = pd.concat(Yl), pd.concat(Pl), pd.DataFrame(SG)
    Y_all.to_csv(S7/f"s7_yearly_stats{sfx}.csv", index=False, encoding="utf-8-sig"); P_all.to_csv(S7/f"s7_period_stats{sfx}.csv", index=False, encoding="utf-8-sig"); S_all.to_csv(S7/f"s7_trend_signs{sfx}.csv", index=False, encoding="utf-8-sig")
    mc = lambda df: (df.scheme == mn["scheme"]) & (df.variant == mn["variant"]) & (df["sample"] == mn["sample"]) & (df.terr_merge == mn["terr_merge"]) & (df.income == mn["income"])
    print("\n[주 구성 기간별 통계]\n", P_all[mc(P_all)][["period"] + SPEC["stats"]].round(4).to_string(index=False))
    rows = []
    for st in SPEC["stats"]:
        d = S_all[S_all.stat == st]
        if not st.startswith("theil"): d = d[d.income == "fixed"]
        b = d[mc(d)].iloc[0]
        for kind in ["sign_trend", "sign_delta"]:
            for samp, g in [("전체 구성", d), ("표본=all", d[d["sample"] == "all"]), ("표본=fixed14", d[d["sample"] == "fixed14"])]:
                rows.append(dict(stat=st, kind=kind, subset=samp, main_sign=int(b[kind]), n=len(g), frac_diff=float((g[kind] != b[kind]).mean()), rho_min=g.rho.min(), rho_max=g.rho.max(),
                                 delta_min=g.delta_first_last.min(), delta_max=g.delta_first_last.max()))
    DS = pd.DataFrame(rows); DS["pass_le_0.25"] = DS.frac_diff <= SPEC["direction_rule"] and True if False else DS.frac_diff <= 0.25
    DS.to_csv(S7/f"s7_direction_split{sfx}.csv", index=False, encoding="utf-8-sig")
    print("\n[방향 갈림 비율: 전체 구성 기준]\n", DS[DS.subset == "전체 구성"].round(3).to_string(index=False))
    print("※ 연도 19개 Spearman은 |rho|<0.46이면 부호가 불안정합니다. 부호 일치와 함께 rho·변화량 범위(위 표)를 같이 보고하세요.")

    # ---------- 부트스트랩(주 구성, 영토 별도, 소득 고정): 기간별 통계와 차이의 구간 ----------
    B = 50 if QUICK else int(SPEC["bootstrap"]["B"]); boot_rows = []
    for sample in ["all", "fixed14"]:
        d = pools[(mn["scheme"], mn["variant"])]
        if sample == "fixed14": d = d[d.venue_short.isin(FIXED)]
        pid_c = d.pid.astype("category"); cc = d.country.astype("category"); cols = np.array(cc.cat.categories)
        X = sp.csr_matrix((d.w.values, (pid_c.cat.codes.values, cc.cat.codes.values)), shape=(len(pid_c.cat.categories), len(cols))); XT = X.T.tocsr()
        py = d.groupby("pid").year.first().reindex(pid_c.cat.categories).values; pmask = {k: ((py >= a) & (py <= b)).astype(float) for k, (a, b) in PER.items()}
        gcol = {k: np.array([grp_of(c, (a+b)//2) for c in cols]) for k, (a, b) in PER.items()}
        def pstats(v):
            out = {}
            for k, (a, b) in PER.items():
                vec = XT @ (v*pmask[k]); st = stats_of(vec, cols, (a+b)//2, "fixed")
                for g in GROUPS: st[f"share_{g}"] = float(vec[gcol[k] == g].sum()/vec.sum())
                out[k] = st
            return out
        obs = pstats(np.ones(X.shape[0])); keys = list(obs["P0709"].keys()); arr = {k: {s: [] for s in keys} for k in PER}
        for _ in range(B):
            r = pstats(rng.poisson(1.0, X.shape[0]).astype(float))
            for k in PER:
                for s in keys: arr[k][s].append(r[k].get(s, np.nan))
        pairs = [("P0709", "P1014"), ("P1014", "P1519"), ("P1519", "P2025"), ("P0709", "P2025")]
        for s in keys:
            for k in PER:
                x = np.array(arr[k][s], float); boot_rows.append(dict(sample=sample, stat=s, what=k, obs=obs[k].get(s), lo=np.nanpercentile(x, 2.5), med=np.nanmedian(x), hi=np.nanpercentile(x, 97.5)))
            for a, b in pairs:
                x = np.array(arr[b][s], float) - np.array(arr[a][s], float); boot_rows.append(dict(sample=sample, stat=s, what=f"{a}->{b}", obs=obs[b].get(s)-obs[a].get(s), lo=np.nanpercentile(x, 2.5), med=np.nanmedian(x), hi=np.nanpercentile(x, 97.5)))
    BT = pd.DataFrame(boot_rows); BT.to_csv(S7/f"s7_bootstrap{sfx}.csv", index=False, encoding="utf-8-sig")
    print(f"\n[부트스트랩 B={B}: 첫~마지막 기간 변화(P0709->P2025), 구간이 0을 포함하지 않는 통계]")
    z = BT[BT.what == "P0709->P2025"]; z = z.assign(excl0=(z.lo > 0) | (z.hi < 0)); print(z[["sample", "stat", "obs", "lo", "hi", "excl0"]].round(4).to_string(index=False))

    # ---------- 신규 국가 진입(창 안, 좌측 절단 표시) ----------
    kn = [c for c in mainG.columns if c not in ("UNK", "ZZ_former")]; ent = []
    for c in kn:
        act = mainG[c][mainG[c] >= AMIN]
        if len(act): fy = int(act.index.min()); ent.append(dict(country=c, first_active_year=fy, left_censored=(fy == 2007), income_fixed=inc_of(c, 2016, "fixed"), total_mass=float(mainG[c].sum())))
    E = pd.DataFrame(ent).sort_values("first_active_year"); E["period"] = E.first_active_year.map(YP); E.to_csv(S7/f"s7_country_entry{sfx}.csv", index=False, encoding="utf-8-sig")
    print("\n[기간별 신규 활성 국가 수(2007 첫 해 제외, 소득군별)]\n", E[~E.left_censored].groupby(["period", "income_fixed"]).size().unstack(fill_value=0).to_string())

    # ---------- 저널 통제: 지역군 몫 변화의 저널 내/간 정확 분해 ----------
    d = add_group(pools[(mn["scheme"], mn["variant"])][["year", "venue_short", "country", "w"]]); d["per"] = d.year.map(YP)
    tab = d.groupby(["per", "venue_short", "g"]).w.sum().unstack(fill_value=0).reindex(columns=GROUPS, fill_value=0); jr = []
    keys_p = list(PER)
    for a, b in zip(keys_p[:-1], keys_p[1:]):
        T0, T1 = tab.loc[a], tab.loc[b]; V = T0.index.union(T1.index); T0, T1 = T0.reindex(V, fill_value=0), T1.reindex(V, fill_value=0)
        n0, n1 = T0.sum(1).values, T1.sum(1).values; w0, w1 = n0/n0.sum(), n1/n1.sum()
        s0 = np.divide(T0.values, n0[:, None], out=np.zeros(T0.shape), where=n0[:, None] > 0); s1 = np.divide(T1.values, n1[:, None], out=np.zeros(T1.shape), where=n1[:, None] > 0)
        wi = (((w0+w1)/2)[:, None]*(s1-s0)).sum(0); bt = (((s0+s1)/2)*(w1-w0)[:, None]).sum(0); tot = (w1[:, None]*s1).sum(0) - (w0[:, None]*s0).sum(0)
        if np.abs(wi+bt-tot).max() > 1e-9: raise Halt("불변식 실패: 저널 내/간 분해")
        for i, g in enumerate(GROUPS): jr.append(dict(pair=f"{a}->{b}", group=g, delta=tot[i], within_journal=wi[i], between_journal=bt[i], between_share=abs(bt[i])/(abs(wi[i])+abs(bt[i])+1e-15)))
    JR = pd.DataFrame(jr); JR.to_csv(S7/f"s7_journal_control{sfx}.csv", index=False, encoding="utf-8-sig"); print("\n[저널 통제: 지역군 몫 변화]\n", JR.round(4).to_string(index=False))

    # ---------- 7.2 저자 층: 신규 저자의 지역 구성 + 지역군 분해 ----------
    F = load_F1X(); pal = pd.read_csv(PAL, usecols=["pid", "author_pos", "name_norm"], dtype={"pid": str}).dropna().drop_duplicates(["pid", "author_pos"])
    first = pal.merge(F[["pid", "year"]], on="pid").groupby("name_norm").year.min()
    A = pd.read_parquet(S1/"authorship_level.parquet") if (S1/"authorship_level.parquet").exists() else pd.read_csv(S1/"authorship_level.csv.gz")
    A["pid"] = A.pid.astype(str); A["countries"] = A.countries.fillna(""); Rr = pd.read_csv(S1/"authorship_level_recovered.csv.gz", dtype={"pid": str}); Rr["countries_rec"] = Rr.countries_rec.fillna("")
    A = A.merge(Rr[["pid", "rank", "countries_rec"]], on=["pid", "rank"], how="left"); A["countries_rec"] = A.countries_rec.fillna("")
    A["cs_main"] = A.countries; A["cs_rec"] = np.where((A.countries == "") & (A.countries_rec != ""), A.countries_rec, A.countries)
    A["n"] = A.groupby("pid")["rank"].transform("size"); A["author_pos"] = A["rank"] + 1
    A = A.merge(pal, on=["pid", "author_pos"], how="left"); print(f"\n저자 행 {len(A):,} | 이름 매칭 실패 {A.name_norm.isna().mean():.2%} (제외, 질량 손실로 보고)"); A = A.dropna(subset=["name_norm"])
    A = A.merge(F[["pid", "year", "venue_short", "has_topic", "topic150", "in_sample"]], on="pid", how="inner")
    tops = np.array(sorted(pd.read_csv(S3/"L7_topic_grade.csv").topic.astype(int))); tix = {t: i for i, t in enumerate(tops)}; K = len(tops)
    A["ti"] = np.where(tf(A.has_topic) & A.topic150.notna(), A.topic150.fillna(-1).astype(int).map(lambda t: tix.get(t, -1)), -1)
    risk = pd.read_csv(DATA/"F2_risk_names.csv"); RISK = set(risk[tf(risk.ambiguous) | tf(risk.heavy)].name_norm)
    gi_map = {g: i for i, g in enumerate(GROUPS)}

    def author_rows(cscol, variant):
        R = A if variant == "none" else A[~A.name_norm.isin(RISK)]
        s = R[cscol].where(R[cscol] != "", "UNK").str.split("|"); k = s.str.len().values
        X = pd.DataFrame({"name": np.repeat(R.name_norm.values, k), "year": np.repeat(R.year.values, k), "ti": np.repeat(R.ti.values, k), "w": np.repeat((1.0/R.n.values)/k, k),
                          "country": np.concatenate(s.values)})
        X = add_group(X); X["gi"] = X.g.map(gi_map).astype(int); X["per"] = X.year.map(YP); return X
    def amat(r, idx, N): return sp.coo_matrix((r.w.values/N, (idx.get_indexer(r.name), r.gi.values)), shape=(len(idx), len(GROUPS))).tocsr()
    def decomp5(M0, M1, en, er):
        w0, w1 = np.asarray(M0.sum(1)).ravel(), np.asarray(M1.sum(1)).ravel(); S0_, S1_ = np.asarray(M0.sum(0)).ravel(), np.asarray(M1.sum(0)).ravel(); Sb = (S0_+S1_)/2
        cont, ex = (w0 > 0) & (w1 > 0), (w0 > 0) & (w1 == 0); C0, C1 = M0[cont].toarray(), M1[cont].toarray(); a0, a1 = w0[cont], w1[cont]; s0, s1 = C0/a0[:, None], C1/a1[:, None]
        tm = dict(adapt=(((a0+a1)/2)[:, None]*(s1-s0)).sum(0), realloc=((((s0+s1)/2)-Sb)*(a1-a0)[:, None]).sum(0),
                  entry_new=np.asarray(M1[en].sum(0)).ravel()-Sb*w1[en].sum(), entry_ret=np.asarray(M1[er].sum(0)).ravel()-Sb*w1[er].sum(),
                  exit=-(np.asarray(M0[ex].sum(0)).ravel()-Sb*w0[ex].sum()))
        if np.abs(sum(tm.values())-(S1_-S0_)).max() > 1e-9: raise Halt("불변식 실패: 지역군 5항 분해")
        return tm, S0_, S1_, w0, w1, cont, ex
    rows_comp, rows_dec, rows_path = [], [], []
    for variant, cscol in [("base", "cs_main"), ("none", "cs_main"), ("base_rec", "cs_rec")]:
        R = author_rows(cscol, "none" if variant == "none" else "base")
        if abs(R.w.sum() - (A if variant == "none" else A[~A.name_norm.isin(RISK)]).eval("1.0/n").sum()) > 1e-6: raise Halt("불변식 실패: 저자 질량 보존")
        for a, b in zip(keys_p[:-1], keys_p[1:]):
            r0, r1 = R[R.per == a], R[R.per == b]; N0, N1 = r0.w.sum(), r1.w.sum(); idx = pd.Index(pd.concat([r0.name, r1.name]).unique())
            M0, M1 = amat(r0, idx, N0), amat(r1, idx, N1); fy = idx.map(first).fillna(9999).values
            w0, w1 = np.asarray(M0.sum(1)).ravel(), np.asarray(M1.sum(1)).ravel(); en_all = (w0 == 0) & (w1 > 0); er = en_all & (fy < PER[b][0]); en = en_all & ~er
            tm, S0_, S1_, w0, w1, cont, ex = decomp5(M0, M1, en, er)
            for i, g in enumerate(GROUPS):
                rows_dec.append(dict(variant=variant, pair=f"{a}->{b}", group=g, share0=S0_[i], share1=S1_[i], delta=S1_[i]-S0_[i], **{k: v[i] for k, v in tm.items()}))
            roles = {"continuing_t1": (M1[cont], w1[cont]), "new_t1": (M1[en], w1[en]), "returning_t1": (M1[er], w1[er]), "exit_t0": (M0[ex], w0[ex])}
            for rn, (Mx, wx) in roles.items():
                v = np.asarray(Mx.sum(0)).ravel(); rows_comp.append(dict(variant=variant, pair=f"{a}->{b}", role=rn, mass=float(wx.sum()), authors=int(Mx.shape[0]), **{f"share_{g}": v[i]/v.sum() for i, g in enumerate(GROUPS)}))
            if variant != "base": continue
            q0, q1 = r1[r1.ti >= 0], None   # 신규 대 지속 저자의 토픽 분포 경로 분해(t1, 지역군 경유)
            role = pd.Series(np.where(~idx.isin([]), "x", "x"), index=idx); cm = pd.Series(cont, index=idx); nm = pd.Series(en, index=idx)
            q = q0.assign(c=cm.reindex(q0.name).values, e=nm.reindex(q0.name).values)
            mk = lambda m: sp.coo_matrix((q[m].w.values, (q[m].gi.values, q[m].ti.values)), shape=(len(GROUPS), K)).toarray()
            Ev, Pv = mk(q.e.values.astype(bool)), mk(q.c.values.astype(bool))
            if Ev.sum() > 0 and Pv.sum() > 0:
                e, p_ = Ev.sum(1)/Ev.sum(), Pv.sum(1)/Pv.sum(); dist = lambda Z: np.divide(Z, Z.sum(1, keepdims=True), out=np.zeros_like(Z), where=Z.sum(1, keepdims=True) > 0)
                Ej, Pj = dist(Ev), dist(Pv); wi_ = (((e+p_)/2)[:, None]*(Ej-Pj)).sum(0); bt_ = (((Ej+Pj)/2)*(e-p_)[:, None]).sum(0); gap = Ev.sum(0)/Ev.sum()-Pv.sum(0)/Pv.sum()
                if np.abs(wi_+bt_-gap).max() > 1e-9: raise Halt("불변식 실패: 신규-지속 지역 경로 분해")
                for j, t in enumerate(tops): rows_path.append(dict(pair=f"{a}->{b}", topic=int(t), gap=gap[j], within_region=wi_[j], between_region=bt_[j]))
    CO, DC, PA = pd.DataFrame(rows_comp), pd.DataFrame(rows_dec), pd.DataFrame(rows_path)
    CO.to_csv(S7/f"s7_entrant_composition{sfx}.csv", index=False, encoding="utf-8-sig"); DC.to_csv(S7/f"s7_author_group_decomp{sfx}.csv", index=False, encoding="utf-8-sig"); PA.to_csv(S7/f"s7_entrant_topic_path{sfx}.csv", index=False, encoding="utf-8-sig")
    print("\n[신규·지속·복귀·퇴출 저자의 지역 몫 (base)]\n", CO[CO.variant == "base"].round(3).to_string(index=False))
    print("\n[지역군 몫 변화의 5항 분해 (base, 지역군별)]\n", DC[DC.variant == "base"].round(4).to_string(index=False))
    print("※ 신규(first_year>=t1 시작)와 복귀(이전 출판)를 분리. 'none' 변형은 위험군 제외 영향, 'base_rec'은 국가 복구 영향(민감도).")
    ab = PA.assign(aw=PA.within_region.abs(), ab=PA.between_region.abs()).groupby("pair")[["aw", "ab"]].sum(); ab["between_abs_share"] = ab.ab/(ab.aw+ab.ab); print("\n[신규-지속 토픽 차이의 지역 간 성분 비중(절대합)]\n", ab.round(3).to_string())

    # ---------- 토픽 변화의 지역 간/내 정확 분해(논문 단위 국가 질량, 두 표본) ----------
    tp = F[tf(F.has_topic) & F.topic150.notna()][["pid", "topic150"]]; tp = tp.assign(ti=tp.topic150.astype(int).map(tix)); print(f"토픽 매핑 실패(제외) {tp.ti.isna().sum():,}편 / {len(tp):,}편"); tp = tp.dropna(subset=["ti"]).assign(ti=lambda x: x.ti.astype(int)); rt = []
    for sample in ["all", "fixed14"]:
        d = pools[(mn["scheme"], mn["variant"])]
        if sample == "fixed14": d = d[d.venue_short.isin(FIXED)]
        d = d.merge(tp[["pid", "ti"]], on="pid", how="inner"); d = add_group(d[["year", "country", "w", "ti"]]); d["per"] = d.year.map(YP); d["gi"] = d.g.map(gi_map).astype(int); d = d.dropna(subset=["per"])
        cube = {k: np.zeros((len(GROUPS), K)) for k in PER}
        for k, g in d.groupby("per"): np.add.at(cube[k], (g.gi.values, g.ti.values), g.w.values)
        for a, b in zip(keys_p[:-1], keys_p[1:]):
            c0, c1 = cube[a], cube[b]; n0, n1 = c0.sum(1), c1.sum(1); w0, w1 = n0/n0.sum(), n1/n1.sum()
            s0 = np.divide(c0, n0[:, None], out=np.zeros_like(c0), where=n0[:, None] > 0); s1 = np.divide(c1, n1[:, None], out=np.zeros_like(c1), where=n1[:, None] > 0)
            wi = ((w0+w1)/2)[:, None]*(s1-s0); bt = ((s0+s1)/2)*(w1-w0)[:, None]; tot = (w1[:, None]*s1).sum(0)-(w0[:, None]*s0).sum(0)
            if np.abs(wi.sum(0)+bt.sum(0)-tot).max() > 1e-9: raise Halt("불변식 실패: 토픽 지역 분해")
            sel = set(np.argsort(-np.abs(tot))[:SPEC["top_change_n"]]) | {tix[t] for t in SPEC["focus_topics"] if t in tix}
            for j in sorted(sel):
                gc = wi[:, j]+bt[:, j]; top = int(np.argmax(np.abs(gc)))
                rt.append(dict(sample=sample, pair=f"{a}->{b}", topic=int(tops[j]), delta=tot[j], within_region=wi[:, j].sum(), between_region=bt[:, j].sum(), between_share=abs(bt[:, j].sum())/(abs(wi[:, j].sum())+abs(bt[:, j].sum())+1e-15),
                               top_group=GROUPS[top], top_group_contrib=gc[top], **{f"c_{g}": gc[i] for i, g in enumerate(GROUPS)}))
    RT = pd.DataFrame(rt); RT.to_csv(S7/f"s7_topic_region_decomp{sfx}.csv", index=False, encoding="utf-8-sig")
    print("\n[초점 토픽의 지역 기여 (표본=all)]\n", RT[(RT["sample"] == "all") & RT.topic.isin(SPEC["focus_topics"])][["pair", "topic", "delta", "within_region", "between_region", "top_group", "top_group_contrib"]].round(4).to_string(index=False))
    print("※ 88·45(전기모터), 69(텐서)가 특정 지역 유입으로 설명되는지는 top_group과 기여 크기로 판정. 인과 아님.")

    # ---------- 기록 ----------
    outs = sorted(f.name for f in S7.glob(f"s7_*{sfx}.csv"))
    meta = dict(spec_sha256=sp_["sha256"], at=now(), quick=QUICK, B=B, outputs={f: sha_f(S7/f) for f in outs}); (S7/f"S7_meta{sfx}.json").write_text(json.dumps(meta, indent=1), encoding="utf-8")
    if not QUICK:
        with (S7/"CHANGELOG_S7.md").open("a", encoding="utf-8") as f:
            f.write(f"- {now()} | S7 본 실행(규칙 해시 {sp_['sha256'][:12]}, 판정 변경 없음): 구성 격자 {len(G_['scheme'])*len(G_['variant'])*len(G_['sample'])*len(G_['terr_merge'])*len(G_['income'])}개, 부트스트랩 B={B}, 저자 층 변형 base/none/base_rec | S7 | S8\n")
        if COMMIT:
            fl = ["paper1/07_S7"] + (["paper1/s7_run.py"] if (ROOT/"paper1/s7_run.py").exists() else []); tag = f"S7-results-{pd.Timestamp.now().strftime('%Y-%m-%d')}"
            if git("tag", "-l", tag).stdout.strip(): raise Halt(f"태그 {tag} 이미 있음")
            r = git("add", "--", *fl);  r.returncode and (_ for _ in ()).throw(Halt(f"git add 실패: {r.stderr.strip()}"))
            r = git("commit", "-m", "S7 results (frozen rules)"); r.returncode and (_ for _ in ()).throw(Halt(f"git commit 실패: {(r.stdout+r.stderr)[:300]}"))
            r = git("tag", "-a", tag, "-m", "S7"); r.returncode and (_ for _ in ()).throw(Halt(f"git tag 실패: {r.stderr.strip()}")); print("커밋·태그:", tag)
    print("\n저장 완료:", len(outs), "개 파일")

def load_F1X():
    F1 = pd.read_csv(DATA/"F1_papers.csv", low_memory=False); F1["pid"] = F1.pid.astype(str)
    exf = S0/"EXCLUDE_PIDS_FROZEN.csv"; log = pd.read_csv(S0/"FREEZE_LOG.csv"); last = log[log.file == exf.name].iloc[-1]
    if sha_f(exf) != last.sha256: raise Halt("EXCLUDE_PIDS_FROZEN.csv 해시 불일치")
    return F1[~F1.pid.isin(set(pd.read_csv(exf, dtype=str, encoding="utf-8-sig").pid))].copy()

try: {"freeze": freeze, "run": run}[MODE]()
except Halt as e: print(f"⛔ 중단: {e}")
````

````python
# -*- coding: utf-8 -*-
# S7·S8 동결 점검 (단일 셀). 결과를 열람하지 않고 구조·해시·스키마만 확인. 읽기 전용(파일 수정 없음).
import json, hashlib, subprocess, re
from pathlib import Path
import pandas as pd

ROOT = Path.home() / "Research/Minnesota_Transport"
DATA, S0, S1, S2, S3, S7, S8 = ROOT/"analysis_data/final", ROOT/"paper1/00_S0", ROOT/"paper1/01_S1", ROOT/"paper1/02_S2", ROOT/"paper1/03_S3", ROOT/"paper1/07_S7", ROOT/"paper1/08_S8"
sha_b = lambda b: hashlib.sha256(b).hexdigest(); sha_f = lambda p: sha_b(Path(p).read_bytes())
spec_hash = lambda s: sha_b(json.dumps({k: v for k, v in s.items() if k not in ("sha256", "created", "refs")}, sort_keys=True, ensure_ascii=False).encode())
git = lambda *a: subprocess.run(["git", "-C", str(ROOT), *a], capture_output=True, text=True)
res = []
def chk(name, ok, detail=""): res.append((("PASS" if ok else "FAIL"), name, detail)); 

# ---- 1. 동결 파일 무결성 (JSON 내용으로 해시 재계산) ----
P = {}
for tag, d, fn in [("S7", S7, "S7_0_PREREG.json"), ("S8", S8, "S8_0_PREREG.json")]:
    p = d/fn; chk(f"{tag} PREREG 존재", p.exists(), str(p))
    chk(f"{tag} CHANGELOG 존재", (d/f"CHANGELOG_{tag}.md").exists())
    if p.exists():
        P[tag] = json.loads(p.read_text(encoding="utf-8"))
        chk(f"{tag} 저장 해시 == 재계산 해시", P[tag]["sha256"] == spec_hash(P[tag]), P[tag]["sha256"][:12])
        chk(f"{tag} seed=2026", P[tag].get("seed") == 2026)

# ---- 2. 참조 입력이 동결 시점과 동일한지 ----
for tag, d in [("S7", S7), ("S8", S8)]:
    if tag not in P: continue
    for name, h in P[tag].get("refs", {}).items():
        cand = [x for x in [S0/name, S1/name, S2/name, S3/name, S7/name] if x.exists()]
        chk(f"{tag} 참조 {name}", bool(cand) and sha_f(cand[0]) == h, "파일 없음" if not cand else "")
# S8이 참조한 S7 PREREG 해시가 현재 S7 PREREG 파일과 같은지
if "S8" in P and "S7_0_PREREG.json" in P["S8"].get("refs", {}):
    chk("S8이 참조한 S7 PREREG 해시 일치", P["S8"]["refs"]["S7_0_PREREG.json"] == sha_f(S7/"S7_0_PREREG.json"))

# ---- 3. 결과 파일이 동결보다 먼저 생기지 않았는지(사전 동결 순서) ----
for tag, d, pat in [("S7", S7, "s7_*.csv"), ("S8", S8, "*.csv")]:
    if tag in P:
        t0 = pd.Timestamp(P[tag]["created"]); early = [f.name for f in d.glob(pat) if pd.Timestamp(f.stat().st_mtime, unit="s") < t0 - pd.Timedelta(seconds=5)]
        chk(f"{tag} 동결 이전에 만든 결과 파일 없음", not early, str(early[:5]))

# ---- 4. S7 입력 스키마 (run 전 사전 점검: 열 이름만 확인) ----
pm = S2/"country_mass_long.parquet"
try:
    M = pd.read_parquet(pm) if pm.exists() else pd.read_csv(S2/"country_mass_long.csv.gz", nrows=1000)
    need = {"pid", "country", "w", "scheme", "variant", "year", "venue_short"}
    chk("country_mass_long 필수 열", need <= set(M.columns), f"누락 {need - set(M.columns)}")
    if need <= set(M.columns):
        chk("scheme 4종 포함", {"equal", "inv_n_authors", "first_last", "full_count"} <= set(pd.read_parquet(pm, columns=["scheme"]).scheme.unique()) if pm.exists() else True)
        vs = set(pd.read_parquet(pm, columns=["variant"]).variant.unique()) if pm.exists() else set(M.variant)
        chk("variant에 main·rec 포함", {"main", "rec"} <= vs, f"실제 {sorted(vs)}")
except Exception as e: chk("country_mass_long 읽기", False, str(e)[:150])
for fn, cols in [(S1/"authorship_level_recovered.csv.gz", {"pid", "rank", "countries_rec"}), (S1/"wb_income_current.csv", {"iso2", "income", "region"}), (S1/"wb_income_history.csv", {"iso2", "year", "income"}),
                 (S1/"title_places.csv", {"pid", "year", "target_countries"}), (ROOT/"analysis_data/paper_author_long.csv", {"pid", "author_pos", "name_norm"}), (DATA/"F2_risk_names.csv", {"name_norm", "ambiguous", "heavy"}),
                 (DATA/"F1_papers.csv", {"pid", "year", "venue_short", "title", "has_topic", "topic150", "in_sample"}), (S3/"L7_topic_grade.csv", {"topic"})]:
    try:
        h = pd.read_csv(fn, nrows=5); chk(f"열 확인 {fn.name}", cols <= set(h.columns), f"누락 {cols - set(h.columns)}")
    except Exception as e: chk(f"열 확인 {fn.name}", False, str(e)[:120])
al = S1/"authorship_level.parquet"
try:
    a = pd.read_parquet(al) if al.exists() else pd.read_csv(S1/"authorship_level.csv.gz", nrows=1000)
    chk("authorship_level 열(pid, rank, countries)", {"pid", "rank", "countries"} <= set(a.columns), f"{list(a.columns)[:8]}")
    chk("rank 최소값 0(= author_pos-1 가정)", int(a["rank"].min()) == 0, f"min={a['rank'].min()}")
except Exception as e: chk("authorship_level 읽기", False, str(e)[:120])
try:
    pa = pd.read_csv(ROOT/"analysis_data/paper_author_long.csv", usecols=["author_pos"], nrows=50000); chk("author_pos 최소값 1", int(pa.author_pos.min()) == 1, f"min={pa.author_pos.min()}")
except Exception as e: chk("author_pos 확인", False, str(e)[:120])
try:
    pre = json.loads((S0/"PREREG.json").read_text(encoding="utf-8")); chk("PREREG.event.fixed_venues 14개", len(pre["event"]["fixed_venues"]) == 14)
except Exception as e: chk("fixed_venues", False, str(e)[:120])
try:
    c = pd.read_csv(S1/"wb_income_current.csv", keep_default_na=False); c = c[(c.income != "INX") & (c.region != "NA")]
    chk("소득군 값 집합", set(c.income) <= {"HIC", "UMC", "LMC", "LIC"}, f"{sorted(set(c.income))}, 국가 {len(c)}개")
except Exception as e: chk("wb_income_current", False, str(e)[:120])

# ---- 5. S8 사전 정규식 점검(컴파일, 규칙 양성 수; 수정 금지, 표본 크기 확인용) ----
if "S8" in P:
    try:
        rx = {k: re.compile(v, re.I) for k, v in P["S8"]["domains"].items()}; chk("S8 도메인 정규식 컴파일", True, f"{len(rx)}종")
        F = pd.read_csv(DATA/"F1_papers.csv", low_memory=False, usecols=["pid", "year", "title"]); F = F[(F.year >= 2007) & F.title.notna()]
        cnt = {k: int(F.title.str.contains(r).sum()) for k, r in rx.items()}
        chk("도메인별 규칙 양성 >= 200편(precision 표본 25편 추출 가능)", min(cnt.values()) >= 200, str(cnt))
        di = P["S8"]["domain_indicators"]; chk("도메인 지표 대응이 각 2종", all(len(v) == 2 for v in di.values()))
        chk("대응된 지표가 정의된 4종 안에 있음", {i for v in di.values() for i in v} <= set(P["S8"]["indicator_defs"]))
    except Exception as e: chk("S8 사전 점검", False, str(e)[:150])

# ---- 6. git 상태 ----
r = git("status", "--short", "--", "paper1/07_S7", "paper1/08_S8"); print("\n[git status (S7, S8)]\n", r.stdout.strip() or "(변경 없음 또는 모두 커밋)")
print("[최근 태그]", git("tag", "--sort=-creatordate").stdout.split()[:6])

R = pd.DataFrame(res, columns=["결과", "항목", "상세"]); pd.set_option("display.width", 250, "display.max_colwidth", 120, "display.max_rows", 200)
print("\n", R.to_string(index=False)); print(f"\nFAIL {int((R.결과 == 'FAIL').sum())}건 / 전체 {len(R)}건")
````

````python
# -*- coding: utf-8 -*-
# S8 대체 지표 후보 검증 (읽기 전용, 결과 미계산). WDI AR5 교통 CO2 + WHO GHO 등록 차량
import json, time, warnings
from pathlib import Path
import numpy as np, pandas as pd, requests
warnings.filterwarnings("ignore"); pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40); pd.set_option("display.max_rows", 200)

ROOT = Path.home() / "Research/Minnesota_Transport"
S1, S2, S8 = ROOT/"paper1/01_S1", ROOT/"paper1/02_S2", ROOT/"paper1/08_S8"
P = json.loads((S8/"S8_0_PREREG.json").read_text(encoding="utf-8")); y0, y1 = P["burden_years"]; sp_ = P["supply"]
cur = pd.read_csv(S1/"wb_income_current.csv", keep_default_na=False); cur = cur[(cur.income != "INX") & (cur.region != "NA")]; valid = set(cur.iso2)

# iso3 -> iso2 (GHO는 iso3 사용)
cj = requests.get("https://api.worldbank.org/v2/country", params=dict(format="json", per_page=400), timeout=60).json()[1]
ISO3 = {c["id"]: c["iso2Code"] for c in cj}

# 공급 질량(커버리지 계산용)
pm = S2/"country_mass_long.parquet"
M = pd.read_parquet(pm) if pm.exists() else pd.read_csv(S2/"country_mass_long.csv.gz", dtype={"pid": str})
M = M[(M.scheme == sp_["scheme"]) & (M.variant == sp_["variant"]) & (M.year >= 2007) & ~M.country.isin(["UNK", "ZZ_former"])]
mass = M.groupby("country").w.sum(); mass = mass[mass.index.isin(valid)]

def summarize(name, d):
    """d: iso2, year, value. 구간 내 커버리지와 공급 질량 커버리지"""
    w = d[d.iso2.isin(valid) & d.year.between(y0, y1)]; ok = set(w.iso2)
    print(f"\n[{name}] 전체 연도 범위 {int(d.year.min()) if len(d) else None}~{int(d.year.max()) if len(d) else None} | 구간 {y0}~{y1} 가용 국가 {len(ok)} | 연도 {sorted(w.year.unique().tolist())} | "
          f"공급 질량 커버리지 {mass[mass.index.isin(ok)].sum()/mass.sum():.3f}")
    if len(w):
        g = w.groupby("iso2").value.mean(); inc = cur.set_index("iso2").income
        t = pd.DataFrame({"v": g, "inc": inc.reindex(g.index)}).dropna()
        print("  소득군별 가용/전체:", {k: f"{(t.inc == k).sum()}/{(cur.income == k).sum()}" for k in ["HIC", "UMC", "LMC", "LIC"]})
        print("  1·2위 국가 비중(소득군별):", {k: (t[t.inc == k].v.sort_values(ascending=False).head(2).div(t[t.inc == k].v.sum()).round(2).to_dict()) for k in ["UMC", "LMC"]})
    return w

# ---------- A. WDI 후보 ----------
for code in ["EN.GHG.CO2.TR.MT.CE.AR5", "EN.GHG.CO2.MT.CE.AR5"]:
    meta = requests.get(f"https://api.worldbank.org/v2/indicator/{code}", params=dict(format="json"), timeout=60).json()
    try: print(f"\n== {code} | 소스 {meta[1][0]['source']['value']} | {meta[1][0]['name']}\n   sourceNote: {meta[1][0].get('sourceNote','')[:400]}")
    except Exception: print(code, "메타 없음")
    rows = []
    for src in [None, 2, 75]:
        rows, page = [], 1
        while True:
            p = dict(format="json", date=f"{y0-3}:{y1+3}", per_page=20000, page=page); src and p.update(source=src)
            j = requests.get(f"https://api.worldbank.org/v2/country/all/indicator/{code}", params=p, timeout=90).json()
            if not isinstance(j, list) or len(j) < 2 or j[1] is None: break
            rows += [(r["country"]["id"], int(r["date"]), r["value"]) for r in j[1] if r["value"] is not None]
            if page >= j[0]["pages"]: break
            page += 1; time.sleep(0.2)
        if rows: print(f"   조회 성공(source={src})"); break
    d = pd.DataFrame(rows, columns=["iso2", "year", "value"])
    if len(d): d.to_csv(S8/f"cand_{code}.csv", index=False); summarize(code, d)
    else: print("   ⛔ 구간 값 없음")

# ---------- B. WHO GHO: 차량 관련 지표 검색 ----------
GHO = "https://ghoapi.azureedge.net/api"
try:
    ind = requests.get(f"{GHO}/Indicator", params={"$filter": "contains(IndicatorName,'vehicle')"}, timeout=90).json()["value"]
    print("\n[GHO 'vehicle' 지표]"); [print(" ", x["IndicatorCode"], "|", x["IndicatorName"][:110]) for x in ind]
    for x in ind:
        if not any(k in x["IndicatorName"].lower() for k in ["registered", "number of", "total"]): continue
        r = requests.get(f"{GHO}/{x['IndicatorCode']}", params={"$filter": "SpatialDimType eq 'COUNTRY'"}, timeout=120).json()["value"]
        if not r: continue
        df = pd.DataFrame(r); df = df[df.NumericValue.notna()]
        df = pd.DataFrame(dict(iso2=df.SpatialDim.map(ISO3), year=df.TimeDim.astype(int), value=df.NumericValue.astype(float), dim1=df.get("Dim1", ""))).dropna(subset=["iso2"])
        print(f"\n  {x['IndicatorCode']} 연도 분포:", df.groupby("year").iso2.nunique().to_dict(), "| Dim1 값:", df.dim1.unique()[:6])
        df.to_csv(S8/f"cand_gho_{x['IndicatorCode']}.csv", index=False); summarize("GHO "+x["IndicatorCode"], df.drop(columns="dim1"))
except Exception as e: print("GHO 조회 실패:", str(e)[:200], "\n→ 브라우저로 https://www.who.int/data/gho 에서 'registered vehicles' 검색 후 csv를 받으세요.")
print("\n※ 값이 구간에 오는지, 총량(대수)인지 비율인지, 연도가 단일인지 확인하세요.")
````

````python
# -*- coding: utf-8 -*-
# =====================================================================================
# S8 공급-부담 정렬 v2 (단일 셀 풀코드). 미실행 코드.
# STEPS 순서: ["freeze"] -> ["tag"] -> ["burden"] -> ["sample"] -> (코더 2명 수작업) -> ["kappa"] x2 -> ["align"]
#  freeze: v1과 동일 항목 검증 후 S8_0_PREREG_v2.json 동결 (지표 코드 교체 + 중국·인도 제외 민감도 사전 명시)
#  tag   : v2 동결 파일·CHANGELOG 커밋과 태그 (freeze 직후, 결과 열람 전에)
#  burden: WB·GHO 입력 수집, 국가 표 구축, 커버리지·인구당 차량 점검 (SBR 미계산)
#  sample: 수작업 표본 300편(coder1/2.csv + coding_master.csv)
#  kappa : 이중 코딩 kappa, 조정 시트, 조정 완료 후 precision/recall(Wilson)와 통과 도메인 확정
#  align : 역량·대상 두 렌즈 SBR, 지표 일관성, GDP 통제 편 순위상관, 중국·인도 제외 민감도
# 해석은 '정렬 간극 지도'까지. 필요 충족 증거 아님. 소속 != 연구 대상, 생태학적 오류 가능.
# =====================================================================================
import json, re, hashlib, warnings, time, subprocess
from pathlib import Path
import numpy as np, pandas as pd, requests
from scipy.stats import rankdata
warnings.filterwarnings("ignore"); pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40); pd.set_option("display.max_rows", 200)

STEPS = ["kappa"]
QUICK = False          # True: 부트스트랩 B=100 (파일명에 _QUICK)
ROOT = Path.home() / "Research/Minnesota_Transport"
DATA, S0, S1, S2, S3, S7, S8 = ROOT/"analysis_data/final", ROOT/"paper1/00_S0", ROOT/"paper1/01_S1", ROOT/"paper1/02_S2", ROOT/"paper1/03_S3", ROOT/"paper1/07_S7", ROOT/"paper1/08_S8"
now = lambda: pd.Timestamp.now().isoformat(timespec="seconds")
sha_b = lambda b: hashlib.sha256(b).hexdigest(); sha_f = lambda p: sha_b(Path(p).read_bytes())
git = lambda *a: subprocess.run(["git", "-C", str(ROOT), *a], capture_output=True, text=True)
class Halt(Exception): pass

# ------------------------- v1과 동일해야 하는 항목(KEEP으로 검증) -------------------------
DOMAINS = {
 "safety": r"\b(safety|crash(es)?|accidents?|fatal\w*|injur\w*|collisions?|casualt\w*|speeding|red[- ]light|vulnerable road user\w*)\b",
 "transit": r"\b(transit|buses|bus|metro|subway|rail|railway|tram|light rail|paratransit|ridership|public transport\w*|brt)\b",
 "electrification_env": r"\b(electric vehicles?|evs?|charging|batter(y|ies)|hybrid|electrif\w*|emissions?|carbon|co2|greenhouse|fuel consumption|energy consumption|air quality|pollut\w*|decarboni\w*|hydrogen|fuel cell)\b",
 "freight": r"\b(freight|logistics?|trucks?|trucking|cargo|goods movement|supply chain|ports?|container\w*|last[- ]mile|warehouse\w*|shipping|intermodal)\b",
 "active_urban": r"\b(walk\w*|pedestrians?|cycl\w*|bicycl\w*|bikes?|micromobility|scooters?|land use|accessibility|urban form|transit[- ]oriented|parking|sidewalks?)\b",
 "infrastructure": r"\b(pavements?|bridges?|asphalt|concrete|road maintenance|highway design|geometric design|structural health|tunnels?|rutting|infrastructure (maintenance|management|asset))\b",
}
DOM_IND = {"safety": ["road_deaths", "vehicles"], "transit": ["urban_pop", "transport_co2"], "electrification_env": ["transport_co2", "vehicles"],
           "freight": ["vehicles", "transport_co2"], "active_urban": ["urban_pop", "road_deaths"], "infrastructure": ["vehicles", "urban_pop"]}
CORE = dict(seed=2026, domains=DOMAINS, domain_indicators=DOM_IND, burden_years=[2014, 2018],
    periods={"P0709": [2007, 2009], "P1014": [2010, 2014], "P1519": [2015, 2019], "P2025": [2020, 2025]},
    supply=dict(scheme="inv_n_authors", variant="main", sample="all", groups="소득군(wb_income_current 고정): HIC, UMC, LMC, LIC. 소득군 없음·미상은 제외", lens_capability="소속 국가 질량(country_mass_long)",
                lens_target="제목 지명(S1 title_places.target_countries), 대상 국가 균등 분할, 지명이 있는 논문만", domain_unit="논문 단위(복수 도메인 허용, 질량은 도메인마다 전량)"),
    sbr="SBR(d,g,t)=도메인 d 공급 몫(그룹 g) / 지표 i 부담 몫(그룹 g), 지표 i 공통 지지 국가(부담 자료가 있는 국가)로 공급·부담 모두 제한. 지표마다 계산",
    country_level=dict(supply_intensity="국가의 도메인 d 질량 / 국가 총 질량", burden_intensity="지표별 강도", control="rank(gdp_pc) 통제 편 순위상관(순위 변환 후 OLS 잔차 상관)", min_country_mass=10.0, bootstrap_B=1000),
    manual=dict(n_random=150, n_per_domain=25, coders=2, kappa_min=0.6, precision_min=0.85, ci="Wilson 95%", rule="도메인별 precision 점추정 >= 0.85인 도메인만 정렬 분석에 사용, 미달 도메인은 부록",
                recall="무작위 150편의 최종 판정 양성 중 규칙 양성 비율(재현율), 통과 기준 아님"),
    direction=dict(rule="셀(도메인, 소득군, 기간)에서 두 지표의 SBR가 1의 같은 쪽(>1 또는 <1)인 비율 >= 0.75이면 '지표 일관'", min_share=0.75),
    interpretation="정렬 간극 지도로만 표현. 소속≠연구 대상, 집계 수준 상관은 생태학적 오류 가능. 두 렌즈가 갈리면 그 자체를 결과로 보고.",
    fallback="precision 미달·지표 불일치 시 RQ5 부록으로 내리고 과학학 계열 저널 전환")
KEEP = list(CORE)

# ------------------------- v2에서 바뀌는 항목 -------------------------
WB = {"pop": "SP.POP.TOTL", "gdp_pc": "NY.GDP.PCAP.KD", "death_rate": "SH.STA.TRAF.P5", "urban_share": "SP.URB.TOTL.IN.ZS", "co2_tr_mt": "EN.GHG.CO2.TR.MT.CE.AR5"}
GHO = {"vehicles_reg": "RS_194"}
SPEC = dict(CORE, version=2,
    decision_status="v1 동결 후 SBR·precision·kappa 미계산 상태에서, 입력 자료 가용성만 근거로 지표 코드 교체(결과 열람 없음). 나머지 사양은 v1과 동일.",
    change_reason="IS.VEH.NVEH.P3는 구 소스(Africa Development Indicators, 2013 갱신), EN.CO2.TRAN.ZS·EN.ATM.CO2E.KT는 WDI Archives로 이동해 2014~2018 자료 없음.",
    wb_codes=WB, gho_codes=GHO,
    burden_rule="WB 지표: 2014~2018 국가별 가용 연도 평균. 차량(GHO RS_194): 구간 내 가용 값(국가당 대부분 1개 연도, 2016 중심)이며 기간 평균이 아님. 결측 국가는 해당 지표에서 제외",
    indicator_defs=dict(road_deaths="death_rate*pop/1e5 (총량)", vehicles="GHO RS_194 등록 차량 수 (총량, 이륜차 포함 여부는 국가별로 다를 수 있음)",
                        urban_pop="urban_share/100*pop (총량)", transport_co2="EN.GHG.CO2.TR.MT.CE.AR5 (Mt CO2e, WDI 정의 교통 부문 총량, 민간항공·수운 포함)"),
    intensity_defs=dict(road_deaths="death_rate", vehicles="vehicles/pop*1e3", urban_pop="urban_share", transport_co2="co2_tr_mt*1e6/pop (t/인)"),
    sensitivity=dict(exclude_country_sets=[["CN"], ["IN"], ["CN", "IN"]], applies_to="소득군 SBR(공급·부담 양쪽에서 해당 국가 제거)과 국가 수준 편 순위상관(국가 제거)",
                     rule="주 결과는 바꾸지 않고, 주 결과와 SBR이 1의 같은 쪽인 셀의 비율을 보고. 사전 명시 분석"))
spec_hash = lambda s: sha_b(json.dumps({k: v for k, v in s.items() if k not in ("sha256", "created", "refs")}, sort_keys=True, ensure_ascii=False).encode())
V1P = S8/"S8_0_PREREG.json"; V2P = S8/"S8_0_PREREG_v2.json"
INT = dict(road_deaths="death_rate", vehicles="veh_per1000", urban_pop="urban_share", transport_co2="co2_tr_pc")   # intensity_defs와 대응

def check_spec():
    if not V2P.exists(): raise Halt("먼저 freeze")
    s = json.loads(V2P.read_text(encoding="utf-8"))
    if s["sha256"] != spec_hash(SPEC) or spec_hash(s) != s["sha256"]: raise Halt("동결 파일과 현재 SPEC이 다릅니다(사후 변경 금지)")
    return s

RX = {k: re.compile(v, re.I) for k, v in DOMAINS.items()}
PER = {k: tuple(v) for k, v in SPEC["periods"].items()}; YP = {y: k for k, (a, b) in PER.items() for y in range(a, b+1)}
def load_F1X():
    F1 = pd.read_csv(DATA/"F1_papers.csv", low_memory=False); F1["pid"] = F1.pid.astype(str)
    exf = S0/"EXCLUDE_PIDS_FROZEN.csv"; log = pd.read_csv(S0/"FREEZE_LOG.csv"); last = log[log.file == exf.name].iloc[-1]
    if sha_f(exf) != last.sha256: raise Halt("EXCLUDE_PIDS_FROZEN.csv 해시 불일치")
    F = F1[~F1.pid.isin(set(pd.read_csv(exf, dtype=str, encoding="utf-8-sig").pid))].copy(); return F[F.year >= 2007].copy()
def flag_domains(titles): return pd.DataFrame({d: titles.fillna("").str.contains(rx).values for d, rx in RX.items()}, index=titles.index)
def wilson(k, n, z=1.96):
    if n == 0: return (np.nan, np.nan)
    p = k/n; d = 1+z*z/n; c = p+z*z/(2*n); h = z*np.sqrt(p*(1-p)/n+z*z/(4*n*n)); return ((c-h)/d, (c+h)/d)
def kappa(a, b):
    a, b = np.asarray(a, bool), np.asarray(b, bool); po = (a == b).mean(); pe = a.mean()*b.mean()+(1-a.mean())*(1-b.mean()); return (po-pe)/(1-pe) if pe < 1 else np.nan
def partial_spearman(x, y, z):
    rx_, ry_, rz = rankdata(x), rankdata(y), rankdata(z); A = np.c_[np.ones(len(z)), rz]
    res = lambda v: v - A @ np.linalg.lstsq(A, v, rcond=None)[0]; a, b = res(rx_), res(ry_); return float(np.corrcoef(a, b)[0, 1])
def load_mass():
    pm = S2/"country_mass_long.parquet"; M = pd.read_parquet(pm) if pm.exists() else pd.read_csv(S2/"country_mass_long.csv.gz", dtype={"pid": str}); M["pid"] = M.pid.astype(str)
    sp_ = SPEC["supply"]; M = M[(M.scheme == sp_["scheme"]) & (M.variant == sp_["variant"]) & (M.year >= 2007)]; return M[~M.country.isin(["UNK", "ZZ_former"])]

# ============================== freeze ==============================
def freeze():
    S8.mkdir(parents=True, exist_ok=True)
    if not V1P.exists(): raise Halt("v1 동결 파일(S8_0_PREREG.json)이 없습니다")
    V1 = json.loads(V1P.read_text(encoding="utf-8"))
    if V1["sha256"] != spec_hash(V1): raise Halt("v1 해시 불일치")
    h = spec_hash(SPEC); print("S8 v2 해시(예정)", h[:12])
    bad = [k for k in KEEP if V1.get(k) != SPEC[k]]
    if bad: raise Halt(f"v1과 달라진 항목(바꾸면 안 됨): {bad}")
    print("v1과 동일 확인:", KEEP)
    if V2P.exists(): print("이미 동결됨:", "동일" if json.loads(V2P.read_text(encoding="utf-8"))["sha256"] == h else "⛔ 다름"); return
    refs = {"S8_0_PREREG_v1.json": sha_f(V1P)}; refs.update({f.name: sha_f(f) for f in [S7/"S7_0_PREREG.json", S1/"wb_income_current.csv", S1/"title_places.csv"] if f.exists()})
    V2P.write_text(json.dumps(dict(SPEC, created=now(), sha256=h, refs=refs), ensure_ascii=False, indent=1), encoding="utf-8")
    with (S8/"CHANGELOG_S8.md").open("a", encoding="utf-8") as f:
        f.write(f"- {now()} | S8 v2 동결(해시 {h[:12]}, v1 {V1['sha256'][:12]} 대체): 지표 코드 교체(차량→GHO RS_194, 교통 CO2→EN.GHG.CO2.TR.MT.CE.AR5, 사망률·도시인구 유지), "
                f"중국·인도 제외 민감도 사전 명시. 사유: 원 코드가 구 소스·아카이브로 이동해 2014~2018 자료 없음. SBR·precision·kappa 미계산 상태에서 입력 가용성만 근거로 교체. 도메인 사전·대응·임계값 불변 | S8 설계 | burden\n")
    print("저장: S8_0_PREREG_v2.json")

def tag():
    check_spec(); t = f"S8-v2-freeze-{pd.Timestamp.now().strftime('%Y-%m-%d')}"
    if git("tag", "-l", t).stdout.strip(): print("이미 있음:", t); return
    r = git("add", "--", "paper1/08_S8/S8_0_PREREG_v2.json", "paper1/08_S8/CHANGELOG_S8.md"); print("add", r.returncode, r.stderr.strip())
    r = git("commit", "-m", "S8 v2 freeze (input indicators only, no results)"); print("commit", r.returncode, (r.stdout + r.stderr)[:200])
    r = git("tag", "-a", t, "-m", "S8 v2 freeze"); print("tag", r.returncode, r.stderr.strip()); print(git("tag", "--sort=-creatordate").stdout.split()[:3])

# ============================== burden ==============================
def burden():
    check_spec(); y0, y1 = SPEC["burden_years"]
    cur = pd.read_csv(S1/"wb_income_current.csv", keep_default_na=False); cur = cur[(cur.income != "INX") & (cur.region != "NA")]; valid = set(cur.iso2); T = {}
    for name, code in WB.items():
        f = S8/f"wb_{name}.csv"
        if not f.exists():
            rows, page = [], 1
            try:
                while True:
                    j = requests.get(f"https://api.worldbank.org/v2/country/all/indicator/{code}", params=dict(format="json", date=f"{y0-3}:{y1+3}", per_page=20000, page=page), timeout=90).json()
                    if not isinstance(j, list) or len(j) < 2 or j[1] is None: print(f"※ {name}({code}) 응답 이상: {str(j)[:150]}"); break
                    rows += [(r["country"]["id"], int(r["date"]), r["value"]) for r in j[1] if r["value"] is not None]
                    if page >= j[0]["pages"]: break
                    page += 1; time.sleep(0.2)
            except Exception as e: print(f"※ {name}({code}) 수집 실패: {str(e)[:120]}")
            if not rows: print(f"⛔ {name}: 값 없음 → 파일 저장 안 함"); continue
            pd.DataFrame(rows, columns=["iso2", "year", "value"]).to_csv(f, index=False)
        d = pd.read_csv(f, keep_default_na=False); d["value"] = pd.to_numeric(d.value, errors="coerce"); d = d[d.iso2.isin(valid) & d.year.between(y0, y1)].dropna(subset=["value"])
        T[name] = d.groupby("iso2").value.mean()
    iso3 = {c["id"]: c["iso2Code"] for c in requests.get("https://api.worldbank.org/v2/country", params=dict(format="json", per_page=400), timeout=60).json()[1]}
    for name, code in GHO.items():
        f = S8/f"gho_{name}.csv"
        if not f.exists():
            df = pd.DataFrame(requests.get(f"https://ghoapi.azureedge.net/api/{code}", params={"$filter": "SpatialDimType eq 'COUNTRY'"}, timeout=120).json().get("value", []))
            if df.empty or "NumericValue" not in df: print(f"⛔ {name}({code}): GHO 값 없음 → 파일 저장 안 함"); continue
            df = df[df.NumericValue.notna()]
            pd.DataFrame(dict(iso2=df.SpatialDim.map(iso3), year=df.TimeDim.astype(int), value=df.NumericValue.astype(float))).dropna(subset=["iso2"]).to_csv(f, index=False)
        d = pd.read_csv(f, keep_default_na=False); d["value"] = pd.to_numeric(d.value, errors="coerce"); d = d[d.iso2.isin(valid) & d.year.between(y0, y1)].dropna(subset=["value"])
        print(f"[GHO {name}] 국가 {d.iso2.nunique()} | 값 2개 이상인 국가 {int((d.groupby('iso2').size() > 1).sum())} | 연도 분포 {d.groupby('year').iso2.nunique().to_dict()}")
        T[name] = d.groupby("iso2").value.mean()
    B = pd.DataFrame(T)
    for n in ["pop", "gdp_pc", "death_rate", "urban_share", "co2_tr_mt", "vehicles_reg"]:
        if n not in B: B[n] = np.nan
    B["road_deaths"] = B.death_rate*B["pop"]/1e5; B["vehicles"] = B.vehicles_reg; B["urban_pop"] = B.urban_share/100*B["pop"]; B["transport_co2"] = B.co2_tr_mt
    B["veh_per1000"] = B.vehicles/B["pop"]*1e3; B["co2_tr_pc"] = B.co2_tr_mt*1e6/B["pop"]
    B["income"] = B.index.map(dict(zip(cur.iso2, cur.income))); B.index.name = "iso2"; B.to_csv(S8/"burden_country_table.csv", encoding="utf-8-sig")
    print("\n[지표 커버리지(국가 수)]\n", B[["pop", "gdp_pc", "death_rate", "urban_share", "co2_tr_mt", "vehicles_reg"]].notna().sum().to_string())
    v = B.veh_per1000.dropna().sort_values()
    print("\n[차량 인구당(1000명) 하위 5, 상위 8, 주요국] — 이륜차 포함 여부 등 정의 차이 점검용")
    print(v.head(5).round(0).to_dict(), "\n", v.tail(8).round(0).to_dict(), "\n", {c: round(float(B.veh_per1000.get(c, np.nan)), 0) for c in ["US", "CN", "IN", "ID", "VN", "TH"]})
    print("\n※ 값이 비현실적이면(예: 인도·베트남이 비정상적으로 높음) GHO RS_194 지표 설명을 확인하세요. 차량은 기간 평균이 아니라 가용 연도 값입니다.")

# ============================== sample ==============================
def sample():
    check_spec(); out = S8/"coding_master.csv"
    if out.exists(): print("이미 있음(덮어쓰지 않음):", out); return
    F = load_F1X(); F = F[F.title.notna()].reset_index(drop=True); fl = flag_domains(F.title); rng = np.random.default_rng(SPEC["seed"]); m = SPEC["manual"]
    ri = rng.choice(len(F), m["n_random"], replace=False); S = [F.iloc[ri].assign(stratum="random")]; used = set(F.pid.iloc[ri])
    for d in DOMAINS:
        c = F[fl[d].values & ~F.pid.isin(used)]; k = min(m["n_per_domain"], len(c)); s = c.sample(k, random_state=SPEC["seed"]); S.append(s.assign(stratum=f"prec_{d}")); used |= set(s.pid)
    X = pd.concat(S)[["pid", "year", "venue_short", "title", "stratum"]].reset_index(drop=True); X = X.join(flag_domains(X.title).add_prefix("rule_"))
    X = X.sample(frac=1, random_state=SPEC["seed"]).reset_index(drop=True); X.to_csv(out, index=False, encoding="utf-8-sig")
    base = X[["pid", "year", "venue_short", "title"]].assign(domains="", note="")
    for c in (1, 2): base.to_csv(S8/f"coder{c}.csv", index=False, encoding="utf-8-sig")
    print(f"{len(X)}편 표본. coder1.csv, coder2.csv(규칙 플래그 없음)만 각 코더에게 배포하세요. coding_master.csv는 숨기세요.")
    print("domains 열: 해당 도메인을 '|'로 구분(", " ".join(DOMAINS), ") 또는 none. 두 코더는 서로 상의하지 않습니다.")

# ============================== kappa ==============================
def kappa_step():
    check_spec(); c1, c2 = (pd.read_csv(S8/f"coder{i}.csv", dtype=str, encoding="utf-8-sig").fillna("") for i in (1, 2))
    if (c1.domains == "").any() or (c2.domains == "").any(): raise Halt("coder 시트에 빈 domains가 있습니다(없으면 none)")
    parse = lambda s: set() if s.strip().lower() == "none" else {x.strip() for x in s.split("|") if x.strip()}
    bad = {x for s in pd.concat([c1.domains, c2.domains]) for x in parse(s)} - set(DOMAINS)
    if bad: raise Halt(f"정의되지 않은 도메인 값: {bad}")
    if list(c1.pid) != list(c2.pid): raise Halt("두 시트의 pid 순서가 다릅니다")
    rows = [dict(domain=d, kappa=kappa([d in parse(s) for s in c1.domains], [d in parse(s) for s in c2.domains]), n_pos1=sum(d in parse(s) for s in c1.domains), n_pos2=sum(d in parse(s) for s in c2.domains)) for d in DOMAINS]
    K = pd.DataFrame(rows); K["kappa_ok"] = K.kappa >= SPEC["manual"]["kappa_min"]; K.to_csv(S8/"kappa.csv", index=False, encoding="utf-8-sig"); print("[도메인별 kappa]\n", K.round(3).to_string(index=False))
    dis = c1[["pid", "title"]].assign(c1=c1.domains, c2=c2.domains); dis = dis[[parse(a) != parse(b) for a, b in zip(dis.c1, dis.c2)]].assign(final=""); fp = S8/"adjudication.csv"
    if not fp.exists(): dis.to_csv(fp, index=False, encoding="utf-8-sig"); print(f"불일치 {len(dis)}편 -> adjudication.csv의 final 열을 합의로 채운 뒤 이 단계를 다시 실행(일치 논문은 자동 확정)")
    adj = pd.read_csv(fp, dtype=str, encoding="utf-8-sig").fillna("")
    if (adj.final == "").any(): print(f"조정 미완 {int((adj.final == '').sum())}편: precision 계산 보류"); return
    bad = {x for s in adj.final for x in parse(s)} - set(DOMAINS)
    if bad: raise Halt(f"adjudication final에 정의되지 않은 도메인: {bad}")
    fin = {p: parse(a) for p, a in zip(c1.pid, c1.domains)}; fin.update({p: parse(f) for p, f in zip(adj.pid, adj.final)})
    M = pd.read_csv(S8/"coding_master.csv", dtype={"pid": str}, encoding="utf-8-sig"); res = []
    for d in DOMAINS:
        pr = M[M.stratum == f"prec_{d}"]; k = sum(d in fin[p] for p in pr.pid); lo, hi = wilson(k, len(pr))
        rd = M[M.stratum == "random"]; pos = [p for p in rd.pid if d in fin[p]]; rk = int(rd[rd.pid.isin(pos)][f"rule_{d}"].sum()); rlo, rhi = wilson(rk, len(pos))
        res.append(dict(domain=d, n_prec=len(pr), precision=k/len(pr) if len(pr) else np.nan, p_lo=lo, p_hi=hi, n_pos_random=len(pos), recall=rk/len(pos) if pos else np.nan, r_lo=rlo, r_hi=rhi,
                        passed=bool(len(pr) and k/len(pr) >= SPEC["manual"]["precision_min"])))
    R = pd.DataFrame(res); R.to_csv(S8/"domain_precision_recall.csv", index=False, encoding="utf-8-sig"); print("\n[도메인 사전 precision/recall]\n", R.round(3).to_string(index=False))
    (S8/"domains_passed.json").write_text(json.dumps(dict(passed=R[R.passed].domain.tolist(), failed=R[~R.passed].domain.tolist(), at=now()), ensure_ascii=False, indent=1), encoding="utf-8")
    print("통과:", R[R.passed].domain.tolist(), "| 미달(부록):", R[~R.passed].domain.tolist(), "| 재현율 표본이 작아 구간이 넓습니다(희귀 도메인).")

# ============================== align ==============================
def align():
    check_spec(); pj = S8/"domains_passed.json"
    if not pj.exists(): raise Halt("kappa 단계(조정 완료 후 precision)를 먼저 끝내세요")
    dom_ok = json.loads(pj.read_text(encoding="utf-8"))["passed"]; print("분석 도메인(precision 통과):", dom_ok)
    if not dom_ok: raise Halt("통과 도메인 없음: 계획서 fallback에 따라 RQ5를 부록으로")
    Bt = pd.read_csv(S8/"burden_country_table.csv", keep_default_na=False, index_col=0); inc = Bt["income"].copy()
    Bt = Bt.drop(columns="income").apply(pd.to_numeric, errors="coerce"); Bt["income"] = inc
    INC = Bt["income"].to_dict(); G4 = ["HIC", "UMC", "LMC", "LIC"]; sfx = "_QUICK" if QUICK else ""
    EXC = [("none", [])] + [("excl_" + "_".join(s), s) for s in SPEC["sensitivity"]["exclude_country_sets"]]

    F = load_F1X(); F = F[F.title.notna()]; fl = flag_domains(F.title); fl["pid"] = F.pid.values; fl = fl[["pid"] + dom_ok]
    Mall = load_mass(); M = Mall.copy(); M["per"] = M.year.map(YP); M = M.merge(fl, on="pid", how="inner")
    TP = pd.read_csv(S1/"title_places.csv", dtype={"pid": str}, keep_default_na=False); TP = TP[TP.target_countries != ""]; TP["tc"] = TP.target_countries.str.split("|"); TP = TP.explode("tc")
    TP["w"] = 1.0/TP.groupby("pid").pid.transform("size"); TP = TP.rename(columns={"tc": "country"})[["pid", "country", "w", "year"]]
    TP["per"] = TP.year.astype(int).map(YP); TP = TP.dropna(subset=["per"]).merge(fl, on="pid", how="inner")

    # ---- 정렬 간극 지도 (렌즈 x 제외집합 x 도메인 x 지표 x 기간 x 소득군) ----
    rows = []
    for lens, D0 in [("capability", M), ("target", TP)]:
        for exn, ex in EXC:
            D = D0[D0.country.isin(Bt.index) & ~D0.country.isin(ex)]
            for d in dom_ok:
                for ind in sorted(set(DOM_IND[d])):
                    sup_c = Bt.index[Bt[ind].notna() & ~Bt.index.isin(ex) & Bt["income"].isin(G4)]
                    Bi = Bt.loc[sup_c, ind].groupby(Bt.loc[sup_c, "income"]).sum().reindex(G4).fillna(0); bsh = Bi/Bi.sum()
                    for per in PER:
                        x = D[(D.per == per) & D[d] & D.country.isin(sup_c)]; Sg = x.groupby(x.country.map(INC)).w.sum().reindex(G4).fillna(0)
                        if Sg.sum() <= 0: continue
                        for g in G4: rows.append(dict(lens=lens, excl=exn, domain=d, indicator=ind, period=per, group=g, supply_share=Sg[g]/Sg.sum(), burden_share=bsh[g],
                                                      sbr=(Sg[g]/Sg.sum())/bsh[g] if bsh[g] > 0 else np.nan, supply_mass=Sg.sum()))
    R = pd.DataFrame(rows); R.to_csv(S8/f"alignment_gap_map{sfx}.csv", index=False, encoding="utf-8-sig"); Rm = R[R.excl == "none"]
    print("\n[정렬 간극 지도: 역량 렌즈, 마지막 기간, 제외 없음 (SBR>1 공급 과잉, <1 과소)]")
    print(Rm[(Rm.lens == "capability") & (Rm.period == "P2025")].pivot_table(index=["domain", "indicator"], columns="group", values="sbr").round(2).to_string())

    # ---- 지표 일관성(제외 없음) ----
    cons = []
    for (lens, d, per, g), x in Rm.groupby(["lens", "domain", "period", "group"]):
        x = x[x.indicator.isin(DOM_IND[d])].dropna(subset=["sbr"])
        if len(x) == 2: cons.append(dict(lens=lens, domain=d, period=per, group=g, same_side=bool((x.sbr > 1).nunique() == 1)))
    C = pd.DataFrame(cons); C.to_csv(S8/f"indicator_consistency{sfx}.csv", index=False, encoding="utf-8-sig")
    cs = C.groupby(["lens", "domain"]).same_side.mean().unstack(0).round(3)
    if "capability" in cs: cs["pass_>=0.75(capability)"] = cs["capability"] >= SPEC["direction"]["min_share"]
    print("\n[도메인별 지표 일관 비율]\n", cs.to_string()); print("전체 일관 비율:", C.groupby("lens").same_side.mean().round(3).to_dict())
    print("※ 지표마다 공통 지지 국가 집합이 달라 일관 비율에 표본 차이가 섞입니다.")

    # ---- 렌즈 일치 ----
    W = Rm.pivot_table(index=["domain", "indicator", "period", "group"], columns="lens", values="sbr").dropna()
    if len(W) and {"capability", "target"} <= set(W.columns): print("두 렌즈가 같은 쪽인 비율:", round(float(((W.capability > 1) == (W.target > 1)).mean()), 3), f"(n={len(W)}; 대상 렌즈는 제목 지명 논문만이라 표본이 작음)")
    print("대상 렌즈 논문 수(도메인별, 제목 지명 보유):", {d: int(TP[d].sum()) for d in dom_ok})

    # ---- 중국·인도 제외 민감도: 주 결과와 SBR이 1의 같은 쪽인 셀 비율 ----
    key = ["lens", "domain", "indicator", "period", "group"]; mm = Rm[key + ["sbr"]].rename(columns={"sbr": "sbr_main"}); sens = []
    for exn, _ in EXC[1:]:
        x = R[R.excl == exn][key + ["sbr"]].merge(mm, on=key).dropna(); x["same"] = (x.sbr > 1) == (x.sbr_main > 1)
        for (lens, d), g in x.groupby(["lens", "domain"]): sens.append(dict(excl=exn, lens=lens, domain=d, n=len(g), same_side_as_main=g.same.mean()))
        sens.append(dict(excl=exn, lens="ALL", domain="ALL", n=len(x), same_side_as_main=x.same.mean()))
    SN = pd.DataFrame(sens); SN.to_csv(S8/f"sensitivity_exclusion{sfx}.csv", index=False, encoding="utf-8-sig"); print("\n[제외 민감도: 주 결과와 같은 쪽인 셀 비율]\n", SN.round(3).to_string(index=False))

    # ---- 국가 수준: GDP 통제 편 순위상관 (2007~2025 전 기간 풀링, 강도) ----
    cl = SPEC["country_level"]; rng = np.random.default_rng(SPEC["seed"]); Bn = 100 if QUICK else cl["bootstrap_B"]; pc = []
    for exn, ex in EXC:
        tot = Mall[~Mall.country.isin(ex)].groupby("country").w.sum(); tot = tot[tot >= cl["min_country_mass"]]
        for d in dom_ok:
            dm = M[M[d] & ~M.country.isin(ex)].groupby("country").w.sum().reindex(tot.index).fillna(0); sup = dm/tot
            for ind in DOM_IND[d]:
                df = pd.DataFrame({"s": sup, "b": Bt[INT[ind]].reindex(tot.index), "z": Bt["gdp_pc"].reindex(tot.index)}).dropna()
                if len(df) < 20: pc.append(dict(excl=exn, domain=d, indicator=ind, n=len(df))); continue
                r = partial_spearman(df.s.values, df.b.values, df.z.values); bs = []
                for _ in range(Bn):
                    i = rng.integers(0, len(df), len(df)); bs.append(partial_spearman(df.s.values[i], df.b.values[i], df.z.values[i]))
                pc.append(dict(excl=exn, domain=d, indicator=ind, n=len(df), raw_rho=float(np.corrcoef(rankdata(df.s), rankdata(df.b))[0, 1]), partial_rho=r, lo=np.nanpercentile(bs, 2.5), hi=np.nanpercentile(bs, 97.5)))
    PC = pd.DataFrame(pc); PC.to_csv(S8/f"country_partial_spearman{sfx}.csv", index=False, encoding="utf-8-sig")
    print("\n[국가 수준 편 순위상관(전 기간 풀링, 제외 없음)]\n", PC[PC.excl == "none"].round(3).to_string(index=False))
    print("\n[제외 민감도: 편 순위상관 부호 유지]\n", PC.pivot_table(index=["domain", "indicator"], columns="excl", values="partial_rho").round(3).to_string())
    print("※ 국가 단위 상관은 생태학적 오류 가능, 인과·필요 충족 증거 아님. 부트스트랩은 국가 단위라 논문 단위 불확실성은 미반영. 두 지표 방향이 일관하고 구간이 0을 배제하는 도메인만 '정렬 단서'로 서술.")

    outs = sorted(f.name for f in S8.glob("*.csv")); (S8/f"S8_meta{sfx}.json").write_text(json.dumps(dict(spec_sha256=spec_hash(SPEC), at=now(), quick=QUICK, outputs={f: sha_f(S8/f) for f in outs}), indent=1), encoding="utf-8")
    if not QUICK:
        with (S8/"CHANGELOG_S8.md").open("a", encoding="utf-8") as f:
            f.write(f"- {now()} | S8 align 실행(v2 규칙 해시 {spec_hash(SPEC)[:12]}, 판정 변경 없음): 통과 도메인 {dom_ok}, 렌즈 2종, 제외 민감도 3종(사전 명시), 지표 일관·편 순위상관 보고. 차량은 가용 연도 값, 교통 CO2는 민간항공·수운 포함 | S8 | 본문\n")
    print("저장 완료")

try:
    for s in STEPS: {"freeze": freeze, "tag": tag, "burden": burden, "sample": sample, "kappa": kappa_step, "align": align}[s]()
except Halt as e: print(f"⛔ 중단: {e}")
````

````python
# -*- coding: utf-8 -*-
# =====================================================================================
# S8 공급-부담 정렬 v2 (단일 셀 풀코드). 미실행 코드.
# STEPS 순서: ["freeze"] -> ["tag"] -> ["burden"] -> ["sample"] -> (코더 2명 수작업) -> ["kappa"] x2 -> ["align"]
#  freeze: v1과 동일 항목 검증 후 S8_0_PREREG_v2.json 동결 (지표 코드 교체 + 중국·인도 제외 민감도 사전 명시)
#  tag   : v2 동결 파일·CHANGELOG 커밋과 태그 (freeze 직후, 결과 열람 전에)
#  burden: WB·GHO 입력 수집, 국가 표 구축, 커버리지·인구당 차량 점검 (SBR 미계산)
#  sample: 수작업 표본 300편(coder1/2.csv + coding_master.csv)
#  kappa : 이중 코딩 kappa, 조정 시트, 조정 완료 후 precision/recall(Wilson)와 통과 도메인 확정
#  align : 역량·대상 두 렌즈 SBR, 지표 일관성, GDP 통제 편 순위상관, 중국·인도 제외 민감도
# 해석은 '정렬 간극 지도'까지. 필요 충족 증거 아님. 소속 != 연구 대상, 생태학적 오류 가능.
# =====================================================================================
import json, re, hashlib, warnings, time, subprocess
from pathlib import Path
import numpy as np, pandas as pd, requests
from scipy.stats import rankdata
warnings.filterwarnings("ignore"); pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40); pd.set_option("display.max_rows", 200)

STEPS = ["align"]
QUICK = False          # True: 부트스트랩 B=100 (파일명에 _QUICK)
ROOT = Path.home() / "Research/Minnesota_Transport"
DATA, S0, S1, S2, S3, S7, S8 = ROOT/"analysis_data/final", ROOT/"paper1/00_S0", ROOT/"paper1/01_S1", ROOT/"paper1/02_S2", ROOT/"paper1/03_S3", ROOT/"paper1/07_S7", ROOT/"paper1/08_S8"
now = lambda: pd.Timestamp.now().isoformat(timespec="seconds")
sha_b = lambda b: hashlib.sha256(b).hexdigest(); sha_f = lambda p: sha_b(Path(p).read_bytes())
git = lambda *a: subprocess.run(["git", "-C", str(ROOT), *a], capture_output=True, text=True)
class Halt(Exception): pass

# ------------------------- v1과 동일해야 하는 항목(KEEP으로 검증) -------------------------
DOMAINS = {
 "safety": r"\b(safety|crash(es)?|accidents?|fatal\w*|injur\w*|collisions?|casualt\w*|speeding|red[- ]light|vulnerable road user\w*)\b",
 "transit": r"\b(transit|buses|bus|metro|subway|rail|railway|tram|light rail|paratransit|ridership|public transport\w*|brt)\b",
 "electrification_env": r"\b(electric vehicles?|evs?|charging|batter(y|ies)|hybrid|electrif\w*|emissions?|carbon|co2|greenhouse|fuel consumption|energy consumption|air quality|pollut\w*|decarboni\w*|hydrogen|fuel cell)\b",
 "freight": r"\b(freight|logistics?|trucks?|trucking|cargo|goods movement|supply chain|ports?|container\w*|last[- ]mile|warehouse\w*|shipping|intermodal)\b",
 "active_urban": r"\b(walk\w*|pedestrians?|cycl\w*|bicycl\w*|bikes?|micromobility|scooters?|land use|accessibility|urban form|transit[- ]oriented|parking|sidewalks?)\b",
 "infrastructure": r"\b(pavements?|bridges?|asphalt|concrete|road maintenance|highway design|geometric design|structural health|tunnels?|rutting|infrastructure (maintenance|management|asset))\b",
}
DOM_IND = {"safety": ["road_deaths", "vehicles"], "transit": ["urban_pop", "transport_co2"], "electrification_env": ["transport_co2", "vehicles"],
           "freight": ["vehicles", "transport_co2"], "active_urban": ["urban_pop", "road_deaths"], "infrastructure": ["vehicles", "urban_pop"]}
CORE = dict(seed=2026, domains=DOMAINS, domain_indicators=DOM_IND, burden_years=[2014, 2018],
    periods={"P0709": [2007, 2009], "P1014": [2010, 2014], "P1519": [2015, 2019], "P2025": [2020, 2025]},
    supply=dict(scheme="inv_n_authors", variant="main", sample="all", groups="소득군(wb_income_current 고정): HIC, UMC, LMC, LIC. 소득군 없음·미상은 제외", lens_capability="소속 국가 질량(country_mass_long)",
                lens_target="제목 지명(S1 title_places.target_countries), 대상 국가 균등 분할, 지명이 있는 논문만", domain_unit="논문 단위(복수 도메인 허용, 질량은 도메인마다 전량)"),
    sbr="SBR(d,g,t)=도메인 d 공급 몫(그룹 g) / 지표 i 부담 몫(그룹 g), 지표 i 공통 지지 국가(부담 자료가 있는 국가)로 공급·부담 모두 제한. 지표마다 계산",
    country_level=dict(supply_intensity="국가의 도메인 d 질량 / 국가 총 질량", burden_intensity="지표별 강도", control="rank(gdp_pc) 통제 편 순위상관(순위 변환 후 OLS 잔차 상관)", min_country_mass=10.0, bootstrap_B=1000),
    manual=dict(n_random=150, n_per_domain=25, coders=2, kappa_min=0.6, precision_min=0.85, ci="Wilson 95%", rule="도메인별 precision 점추정 >= 0.85인 도메인만 정렬 분석에 사용, 미달 도메인은 부록",
                recall="무작위 150편의 최종 판정 양성 중 규칙 양성 비율(재현율), 통과 기준 아님"),
    direction=dict(rule="셀(도메인, 소득군, 기간)에서 두 지표의 SBR가 1의 같은 쪽(>1 또는 <1)인 비율 >= 0.75이면 '지표 일관'", min_share=0.75),
    interpretation="정렬 간극 지도로만 표현. 소속≠연구 대상, 집계 수준 상관은 생태학적 오류 가능. 두 렌즈가 갈리면 그 자체를 결과로 보고.",
    fallback="precision 미달·지표 불일치 시 RQ5 부록으로 내리고 과학학 계열 저널 전환")
KEEP = list(CORE)

# ------------------------- v2에서 바뀌는 항목 -------------------------
WB = {"pop": "SP.POP.TOTL", "gdp_pc": "NY.GDP.PCAP.KD", "death_rate": "SH.STA.TRAF.P5", "urban_share": "SP.URB.TOTL.IN.ZS", "co2_tr_mt": "EN.GHG.CO2.TR.MT.CE.AR5"}
GHO = {"vehicles_reg": "RS_194"}
SPEC = dict(CORE, version=2,
    decision_status="v1 동결 후 SBR·precision·kappa 미계산 상태에서, 입력 자료 가용성만 근거로 지표 코드 교체(결과 열람 없음). 나머지 사양은 v1과 동일.",
    change_reason="IS.VEH.NVEH.P3는 구 소스(Africa Development Indicators, 2013 갱신), EN.CO2.TRAN.ZS·EN.ATM.CO2E.KT는 WDI Archives로 이동해 2014~2018 자료 없음.",
    wb_codes=WB, gho_codes=GHO,
    burden_rule="WB 지표: 2014~2018 국가별 가용 연도 평균. 차량(GHO RS_194): 구간 내 가용 값(국가당 대부분 1개 연도, 2016 중심)이며 기간 평균이 아님. 결측 국가는 해당 지표에서 제외",
    indicator_defs=dict(road_deaths="death_rate*pop/1e5 (총량)", vehicles="GHO RS_194 등록 차량 수 (총량, 이륜차 포함 여부는 국가별로 다를 수 있음)",
                        urban_pop="urban_share/100*pop (총량)", transport_co2="EN.GHG.CO2.TR.MT.CE.AR5 (Mt CO2e, WDI 정의 교통 부문 총량, 민간항공·수운 포함)"),
    intensity_defs=dict(road_deaths="death_rate", vehicles="vehicles/pop*1e3", urban_pop="urban_share", transport_co2="co2_tr_mt*1e6/pop (t/인)"),
    sensitivity=dict(exclude_country_sets=[["CN"], ["IN"], ["CN", "IN"]], applies_to="소득군 SBR(공급·부담 양쪽에서 해당 국가 제거)과 국가 수준 편 순위상관(국가 제거)",
                     rule="주 결과는 바꾸지 않고, 주 결과와 SBR이 1의 같은 쪽인 셀의 비율을 보고. 사전 명시 분석"))
spec_hash = lambda s: sha_b(json.dumps({k: v for k, v in s.items() if k not in ("sha256", "created", "refs")}, sort_keys=True, ensure_ascii=False).encode())
V1P = S8/"S8_0_PREREG.json"; V2P = S8/"S8_0_PREREG_v2.json"
INT = dict(road_deaths="death_rate", vehicles="veh_per1000", urban_pop="urban_share", transport_co2="co2_tr_pc")   # intensity_defs와 대응

def check_spec():
    if not V2P.exists(): raise Halt("먼저 freeze")
    s = json.loads(V2P.read_text(encoding="utf-8"))
    if s["sha256"] != spec_hash(SPEC) or spec_hash(s) != s["sha256"]: raise Halt("동결 파일과 현재 SPEC이 다릅니다(사후 변경 금지)")
    return s

RX = {k: re.compile(v, re.I) for k, v in DOMAINS.items()}
PER = {k: tuple(v) for k, v in SPEC["periods"].items()}; YP = {y: k for k, (a, b) in PER.items() for y in range(a, b+1)}
def load_F1X():
    F1 = pd.read_csv(DATA/"F1_papers.csv", low_memory=False); F1["pid"] = F1.pid.astype(str)
    exf = S0/"EXCLUDE_PIDS_FROZEN.csv"; log = pd.read_csv(S0/"FREEZE_LOG.csv"); last = log[log.file == exf.name].iloc[-1]
    if sha_f(exf) != last.sha256: raise Halt("EXCLUDE_PIDS_FROZEN.csv 해시 불일치")
    F = F1[~F1.pid.isin(set(pd.read_csv(exf, dtype=str, encoding="utf-8-sig").pid))].copy(); return F[F.year >= 2007].copy()
def flag_domains(titles): return pd.DataFrame({d: titles.fillna("").str.contains(rx).values for d, rx in RX.items()}, index=titles.index)
def wilson(k, n, z=1.96):
    if n == 0: return (np.nan, np.nan)
    p = k/n; d = 1+z*z/n; c = p+z*z/(2*n); h = z*np.sqrt(p*(1-p)/n+z*z/(4*n*n)); return ((c-h)/d, (c+h)/d)
def kappa(a, b):
    a, b = np.asarray(a, bool), np.asarray(b, bool); po = (a == b).mean(); pe = a.mean()*b.mean()+(1-a.mean())*(1-b.mean()); return (po-pe)/(1-pe) if pe < 1 else np.nan
def partial_spearman(x, y, z):
    rx_, ry_, rz = rankdata(x), rankdata(y), rankdata(z); A = np.c_[np.ones(len(z)), rz]
    res = lambda v: v - A @ np.linalg.lstsq(A, v, rcond=None)[0]; a, b = res(rx_), res(ry_); return float(np.corrcoef(a, b)[0, 1])
def load_mass():
    pm = S2/"country_mass_long.parquet"; M = pd.read_parquet(pm) if pm.exists() else pd.read_csv(S2/"country_mass_long.csv.gz", dtype={"pid": str}); M["pid"] = M.pid.astype(str)
    sp_ = SPEC["supply"]; M = M[(M.scheme == sp_["scheme"]) & (M.variant == sp_["variant"]) & (M.year >= 2007)]; return M[~M.country.isin(["UNK", "ZZ_former"])]

# ============================== freeze ==============================
def freeze():
    S8.mkdir(parents=True, exist_ok=True)
    if not V1P.exists(): raise Halt("v1 동결 파일(S8_0_PREREG.json)이 없습니다")
    V1 = json.loads(V1P.read_text(encoding="utf-8"))
    if V1["sha256"] != spec_hash(V1): raise Halt("v1 해시 불일치")
    h = spec_hash(SPEC); print("S8 v2 해시(예정)", h[:12])
    bad = [k for k in KEEP if V1.get(k) != SPEC[k]]
    if bad: raise Halt(f"v1과 달라진 항목(바꾸면 안 됨): {bad}")
    print("v1과 동일 확인:", KEEP)
    if V2P.exists(): print("이미 동결됨:", "동일" if json.loads(V2P.read_text(encoding="utf-8"))["sha256"] == h else "⛔ 다름"); return
    refs = {"S8_0_PREREG_v1.json": sha_f(V1P)}; refs.update({f.name: sha_f(f) for f in [S7/"S7_0_PREREG.json", S1/"wb_income_current.csv", S1/"title_places.csv"] if f.exists()})
    V2P.write_text(json.dumps(dict(SPEC, created=now(), sha256=h, refs=refs), ensure_ascii=False, indent=1), encoding="utf-8")
    with (S8/"CHANGELOG_S8.md").open("a", encoding="utf-8") as f:
        f.write(f"- {now()} | S8 v2 동결(해시 {h[:12]}, v1 {V1['sha256'][:12]} 대체): 지표 코드 교체(차량→GHO RS_194, 교통 CO2→EN.GHG.CO2.TR.MT.CE.AR5, 사망률·도시인구 유지), "
                f"중국·인도 제외 민감도 사전 명시. 사유: 원 코드가 구 소스·아카이브로 이동해 2014~2018 자료 없음. SBR·precision·kappa 미계산 상태에서 입력 가용성만 근거로 교체. 도메인 사전·대응·임계값 불변 | S8 설계 | burden\n")
    print("저장: S8_0_PREREG_v2.json")

def tag():
    check_spec(); t = f"S8-v2-freeze-{pd.Timestamp.now().strftime('%Y-%m-%d')}"
    if git("tag", "-l", t).stdout.strip(): print("이미 있음:", t); return
    r = git("add", "--", "paper1/08_S8/S8_0_PREREG_v2.json", "paper1/08_S8/CHANGELOG_S8.md"); print("add", r.returncode, r.stderr.strip())
    r = git("commit", "-m", "S8 v2 freeze (input indicators only, no results)"); print("commit", r.returncode, (r.stdout + r.stderr)[:200])
    r = git("tag", "-a", t, "-m", "S8 v2 freeze"); print("tag", r.returncode, r.stderr.strip()); print(git("tag", "--sort=-creatordate").stdout.split()[:3])

# ============================== burden ==============================
def burden():
    check_spec(); y0, y1 = SPEC["burden_years"]
    cur = pd.read_csv(S1/"wb_income_current.csv", keep_default_na=False); cur = cur[(cur.income != "INX") & (cur.region != "NA")]; valid = set(cur.iso2); T = {}
    for name, code in WB.items():
        f = S8/f"wb_{name}.csv"
        if not f.exists():
            rows, page = [], 1
            try:
                while True:
                    j = requests.get(f"https://api.worldbank.org/v2/country/all/indicator/{code}", params=dict(format="json", date=f"{y0-3}:{y1+3}", per_page=20000, page=page), timeout=90).json()
                    if not isinstance(j, list) or len(j) < 2 or j[1] is None: print(f"※ {name}({code}) 응답 이상: {str(j)[:150]}"); break
                    rows += [(r["country"]["id"], int(r["date"]), r["value"]) for r in j[1] if r["value"] is not None]
                    if page >= j[0]["pages"]: break
                    page += 1; time.sleep(0.2)
            except Exception as e: print(f"※ {name}({code}) 수집 실패: {str(e)[:120]}")
            if not rows: print(f"⛔ {name}: 값 없음 → 파일 저장 안 함"); continue
            pd.DataFrame(rows, columns=["iso2", "year", "value"]).to_csv(f, index=False)
        d = pd.read_csv(f, keep_default_na=False); d["value"] = pd.to_numeric(d.value, errors="coerce"); d = d[d.iso2.isin(valid) & d.year.between(y0, y1)].dropna(subset=["value"])
        T[name] = d.groupby("iso2").value.mean()
    iso3 = {c["id"]: c["iso2Code"] for c in requests.get("https://api.worldbank.org/v2/country", params=dict(format="json", per_page=400), timeout=60).json()[1]}
    for name, code in GHO.items():
        f = S8/f"gho_{name}.csv"
        if not f.exists():
            df = pd.DataFrame(requests.get(f"https://ghoapi.azureedge.net/api/{code}", params={"$filter": "SpatialDimType eq 'COUNTRY'"}, timeout=120).json().get("value", []))
            if df.empty or "NumericValue" not in df: print(f"⛔ {name}({code}): GHO 값 없음 → 파일 저장 안 함"); continue
            df = df[df.NumericValue.notna()]
            pd.DataFrame(dict(iso2=df.SpatialDim.map(iso3), year=df.TimeDim.astype(int), value=df.NumericValue.astype(float))).dropna(subset=["iso2"]).to_csv(f, index=False)
        d = pd.read_csv(f, keep_default_na=False); d["value"] = pd.to_numeric(d.value, errors="coerce"); d = d[d.iso2.isin(valid) & d.year.between(y0, y1)].dropna(subset=["value"])
        print(f"[GHO {name}] 국가 {d.iso2.nunique()} | 값 2개 이상인 국가 {int((d.groupby('iso2').size() > 1).sum())} | 연도 분포 {d.groupby('year').iso2.nunique().to_dict()}")
        T[name] = d.groupby("iso2").value.mean()
    B = pd.DataFrame(T)
    for n in ["pop", "gdp_pc", "death_rate", "urban_share", "co2_tr_mt", "vehicles_reg"]:
        if n not in B: B[n] = np.nan
    B["road_deaths"] = B.death_rate*B["pop"]/1e5; B["vehicles"] = B.vehicles_reg; B["urban_pop"] = B.urban_share/100*B["pop"]; B["transport_co2"] = B.co2_tr_mt
    B["veh_per1000"] = B.vehicles/B["pop"]*1e3; B["co2_tr_pc"] = B.co2_tr_mt*1e6/B["pop"]
    B["income"] = B.index.map(dict(zip(cur.iso2, cur.income))); B.index.name = "iso2"; B.to_csv(S8/"burden_country_table.csv", encoding="utf-8-sig")
    print("\n[지표 커버리지(국가 수)]\n", B[["pop", "gdp_pc", "death_rate", "urban_share", "co2_tr_mt", "vehicles_reg"]].notna().sum().to_string())
    v = B.veh_per1000.dropna().sort_values()
    print("\n[차량 인구당(1000명) 하위 5, 상위 8, 주요국] — 이륜차 포함 여부 등 정의 차이 점검용")
    print(v.head(5).round(0).to_dict(), "\n", v.tail(8).round(0).to_dict(), "\n", {c: round(float(B.veh_per1000.get(c, np.nan)), 0) for c in ["US", "CN", "IN", "ID", "VN", "TH"]})
    print("\n※ 값이 비현실적이면(예: 인도·베트남이 비정상적으로 높음) GHO RS_194 지표 설명을 확인하세요. 차량은 기간 평균이 아니라 가용 연도 값입니다.")

# ============================== sample ==============================
def sample():
    check_spec(); out = S8/"coding_master.csv"
    if out.exists(): print("이미 있음(덮어쓰지 않음):", out); return
    F = load_F1X(); F = F[F.title.notna()].reset_index(drop=True); fl = flag_domains(F.title); rng = np.random.default_rng(SPEC["seed"]); m = SPEC["manual"]
    ri = rng.choice(len(F), m["n_random"], replace=False); S = [F.iloc[ri].assign(stratum="random")]; used = set(F.pid.iloc[ri])
    for d in DOMAINS:
        c = F[fl[d].values & ~F.pid.isin(used)]; k = min(m["n_per_domain"], len(c)); s = c.sample(k, random_state=SPEC["seed"]); S.append(s.assign(stratum=f"prec_{d}")); used |= set(s.pid)
    X = pd.concat(S)[["pid", "year", "venue_short", "title", "stratum"]].reset_index(drop=True); X = X.join(flag_domains(X.title).add_prefix("rule_"))
    X = X.sample(frac=1, random_state=SPEC["seed"]).reset_index(drop=True); X.to_csv(out, index=False, encoding="utf-8-sig")
    base = X[["pid", "year", "venue_short", "title"]].assign(domains="", note="")
    for c in (1, 2): base.to_csv(S8/f"coder{c}.csv", index=False, encoding="utf-8-sig")
    print(f"{len(X)}편 표본. coder1.csv, coder2.csv(규칙 플래그 없음)만 각 코더에게 배포하세요. coding_master.csv는 숨기세요.")
    print("domains 열: 해당 도메인을 '|'로 구분(", " ".join(DOMAINS), ") 또는 none. 두 코더는 서로 상의하지 않습니다.")

# ============================== kappa ==============================
def kappa_step():
    check_spec(); c1, c2 = (pd.read_csv(S8/f"coder{i}.csv", dtype=str, encoding="utf-8-sig").fillna("") for i in (1, 2))
    if (c1.domains == "").any() or (c2.domains == "").any(): raise Halt("coder 시트에 빈 domains가 있습니다(없으면 none)")
    parse = lambda s: set() if s.strip().lower() == "none" else {x.strip() for x in s.split("|") if x.strip()}
    bad = {x for s in pd.concat([c1.domains, c2.domains]) for x in parse(s)} - set(DOMAINS)
    if bad: raise Halt(f"정의되지 않은 도메인 값: {bad}")
    if list(c1.pid) != list(c2.pid): raise Halt("두 시트의 pid 순서가 다릅니다")
    rows = [dict(domain=d, kappa=kappa([d in parse(s) for s in c1.domains], [d in parse(s) for s in c2.domains]), n_pos1=sum(d in parse(s) for s in c1.domains), n_pos2=sum(d in parse(s) for s in c2.domains)) for d in DOMAINS]
    K = pd.DataFrame(rows); K["kappa_ok"] = K.kappa >= SPEC["manual"]["kappa_min"]; K.to_csv(S8/"kappa.csv", index=False, encoding="utf-8-sig"); print("[도메인별 kappa]\n", K.round(3).to_string(index=False))
    dis = c1[["pid", "title"]].assign(c1=c1.domains, c2=c2.domains); dis = dis[[parse(a) != parse(b) for a, b in zip(dis.c1, dis.c2)]].assign(final=""); fp = S8/"adjudication.csv"
    if not fp.exists(): dis.to_csv(fp, index=False, encoding="utf-8-sig"); print(f"불일치 {len(dis)}편 -> adjudication.csv의 final 열을 합의로 채운 뒤 이 단계를 다시 실행(일치 논문은 자동 확정)")
    adj = pd.read_csv(fp, dtype=str, encoding="utf-8-sig").fillna("")
    if (adj.final == "").any(): print(f"조정 미완 {int((adj.final == '').sum())}편: precision 계산 보류"); return
    bad = {x for s in adj.final for x in parse(s)} - set(DOMAINS)
    if bad: raise Halt(f"adjudication final에 정의되지 않은 도메인: {bad}")
    fin = {p: parse(a) for p, a in zip(c1.pid, c1.domains)}; fin.update({p: parse(f) for p, f in zip(adj.pid, adj.final)})
    M = pd.read_csv(S8/"coding_master.csv", dtype={"pid": str}, encoding="utf-8-sig"); res = []
    for d in DOMAINS:
        pr = M[M.stratum == f"prec_{d}"]; k = sum(d in fin[p] for p in pr.pid); lo, hi = wilson(k, len(pr))
        rd = M[M.stratum == "random"]; pos = [p for p in rd.pid if d in fin[p]]; rk = int(rd[rd.pid.isin(pos)][f"rule_{d}"].sum()); rlo, rhi = wilson(rk, len(pos))
        res.append(dict(domain=d, n_prec=len(pr), precision=k/len(pr) if len(pr) else np.nan, p_lo=lo, p_hi=hi, n_pos_random=len(pos), recall=rk/len(pos) if pos else np.nan, r_lo=rlo, r_hi=rhi,
                        passed=bool(len(pr) and k/len(pr) >= SPEC["manual"]["precision_min"])))
    R = pd.DataFrame(res); R.to_csv(S8/"domain_precision_recall.csv", index=False, encoding="utf-8-sig"); print("\n[도메인 사전 precision/recall]\n", R.round(3).to_string(index=False))
    (S8/"domains_passed.json").write_text(json.dumps(dict(passed=R[R.passed].domain.tolist(), failed=R[~R.passed].domain.tolist(), at=now()), ensure_ascii=False, indent=1), encoding="utf-8")
    print("통과:", R[R.passed].domain.tolist(), "| 미달(부록):", R[~R.passed].domain.tolist(), "| 재현율 표본이 작아 구간이 넓습니다(희귀 도메인).")

# ============================== align ==============================
def align():
    check_spec(); pj = S8/"domains_passed.json"
    if not pj.exists(): raise Halt("kappa 단계(조정 완료 후 precision)를 먼저 끝내세요")
    dom_ok = json.loads(pj.read_text(encoding="utf-8"))["passed"]; print("분석 도메인(precision 통과):", dom_ok)
    if not dom_ok: raise Halt("통과 도메인 없음: 계획서 fallback에 따라 RQ5를 부록으로")
    Bt = pd.read_csv(S8/"burden_country_table.csv", keep_default_na=False, index_col=0); inc = Bt["income"].copy()
    Bt = Bt.drop(columns="income").apply(pd.to_numeric, errors="coerce"); Bt["income"] = inc
    INC = Bt["income"].to_dict(); G4 = ["HIC", "UMC", "LMC", "LIC"]; sfx = "_QUICK" if QUICK else ""
    EXC = [("none", [])] + [("excl_" + "_".join(s), s) for s in SPEC["sensitivity"]["exclude_country_sets"]]

    F = load_F1X(); F = F[F.title.notna()]; fl = flag_domains(F.title); fl["pid"] = F.pid.values; fl = fl[["pid"] + dom_ok]
    Mall = load_mass(); M = Mall.copy(); M["per"] = M.year.map(YP); M = M.merge(fl, on="pid", how="inner")
    TP = pd.read_csv(S1/"title_places.csv", dtype={"pid": str}, keep_default_na=False); TP = TP[TP.target_countries != ""]; TP["tc"] = TP.target_countries.str.split("|"); TP = TP.explode("tc")
    TP["w"] = 1.0/TP.groupby("pid").pid.transform("size"); TP = TP.rename(columns={"tc": "country"})[["pid", "country", "w", "year"]]
    TP["per"] = TP.year.astype(int).map(YP); TP = TP.dropna(subset=["per"]).merge(fl, on="pid", how="inner")

    # ---- 정렬 간극 지도 (렌즈 x 제외집합 x 도메인 x 지표 x 기간 x 소득군) ----
    rows = []
    for lens, D0 in [("capability", M), ("target", TP)]:
        for exn, ex in EXC:
            D = D0[D0.country.isin(Bt.index) & ~D0.country.isin(ex)]
            for d in dom_ok:
                for ind in sorted(set(DOM_IND[d])):
                    sup_c = Bt.index[Bt[ind].notna() & ~Bt.index.isin(ex) & Bt["income"].isin(G4)]
                    Bi = Bt.loc[sup_c, ind].groupby(Bt.loc[sup_c, "income"]).sum().reindex(G4).fillna(0); bsh = Bi/Bi.sum()
                    for per in PER:
                        x = D[(D.per == per) & D[d] & D.country.isin(sup_c)]; Sg = x.groupby(x.country.map(INC)).w.sum().reindex(G4).fillna(0)
                        if Sg.sum() <= 0: continue
                        for g in G4: rows.append(dict(lens=lens, excl=exn, domain=d, indicator=ind, period=per, group=g, supply_share=Sg[g]/Sg.sum(), burden_share=bsh[g],
                                                      sbr=(Sg[g]/Sg.sum())/bsh[g] if bsh[g] > 0 else np.nan, supply_mass=Sg.sum()))
    R = pd.DataFrame(rows); R.to_csv(S8/f"alignment_gap_map{sfx}.csv", index=False, encoding="utf-8-sig"); Rm = R[R.excl == "none"]
    print("\n[정렬 간극 지도: 역량 렌즈, 마지막 기간, 제외 없음 (SBR>1 공급 과잉, <1 과소)]")
    print(Rm[(Rm.lens == "capability") & (Rm.period == "P2025")].pivot_table(index=["domain", "indicator"], columns="group", values="sbr").round(2).to_string())

    # ---- 지표 일관성(제외 없음) ----
    cons = []
    for (lens, d, per, g), x in Rm.groupby(["lens", "domain", "period", "group"]):
        x = x[x.indicator.isin(DOM_IND[d])].dropna(subset=["sbr"])
        if len(x) == 2: cons.append(dict(lens=lens, domain=d, period=per, group=g, same_side=bool((x.sbr > 1).nunique() == 1)))
    C = pd.DataFrame(cons); C.to_csv(S8/f"indicator_consistency{sfx}.csv", index=False, encoding="utf-8-sig")
    cs = C.groupby(["lens", "domain"]).same_side.mean().unstack(0).round(3)
    if "capability" in cs: cs["pass_>=0.75(capability)"] = cs["capability"] >= SPEC["direction"]["min_share"]
    print("\n[도메인별 지표 일관 비율]\n", cs.to_string()); print("전체 일관 비율:", C.groupby("lens").same_side.mean().round(3).to_dict())
    print("※ 지표마다 공통 지지 국가 집합이 달라 일관 비율에 표본 차이가 섞입니다.")

    # ---- 렌즈 일치 ----
    W = Rm.pivot_table(index=["domain", "indicator", "period", "group"], columns="lens", values="sbr").dropna()
    if len(W) and {"capability", "target"} <= set(W.columns): print("두 렌즈가 같은 쪽인 비율:", round(float(((W.capability > 1) == (W.target > 1)).mean()), 3), f"(n={len(W)}; 대상 렌즈는 제목 지명 논문만이라 표본이 작음)")
    print("대상 렌즈 논문 수(도메인별, 제목 지명 보유):", {d: int(TP[d].sum()) for d in dom_ok})

    # ---- 중국·인도 제외 민감도: 주 결과와 SBR이 1의 같은 쪽인 셀 비율 ----
    key = ["lens", "domain", "indicator", "period", "group"]; mm = Rm[key + ["sbr"]].rename(columns={"sbr": "sbr_main"}); sens = []
    for exn, _ in EXC[1:]:
        x = R[R.excl == exn][key + ["sbr"]].merge(mm, on=key).dropna(); x["same"] = (x.sbr > 1) == (x.sbr_main > 1)
        for (lens, d), g in x.groupby(["lens", "domain"]): sens.append(dict(excl=exn, lens=lens, domain=d, n=len(g), same_side_as_main=g.same.mean()))
        sens.append(dict(excl=exn, lens="ALL", domain="ALL", n=len(x), same_side_as_main=x.same.mean()))
    SN = pd.DataFrame(sens); SN.to_csv(S8/f"sensitivity_exclusion{sfx}.csv", index=False, encoding="utf-8-sig"); print("\n[제외 민감도: 주 결과와 같은 쪽인 셀 비율]\n", SN.round(3).to_string(index=False))

    # ---- 국가 수준: GDP 통제 편 순위상관 (2007~2025 전 기간 풀링, 강도) ----
    cl = SPEC["country_level"]; rng = np.random.default_rng(SPEC["seed"]); Bn = 100 if QUICK else cl["bootstrap_B"]; pc = []
    for exn, ex in EXC:
        tot = Mall[~Mall.country.isin(ex)].groupby("country").w.sum(); tot = tot[tot >= cl["min_country_mass"]]
        for d in dom_ok:
            dm = M[M[d] & ~M.country.isin(ex)].groupby("country").w.sum().reindex(tot.index).fillna(0); sup = dm/tot
            for ind in DOM_IND[d]:
                df = pd.DataFrame({"s": sup, "b": Bt[INT[ind]].reindex(tot.index), "z": Bt["gdp_pc"].reindex(tot.index)}).dropna()
                if len(df) < 20: pc.append(dict(excl=exn, domain=d, indicator=ind, n=len(df))); continue
                r = partial_spearman(df.s.values, df.b.values, df.z.values); bs = []
                for _ in range(Bn):
                    i = rng.integers(0, len(df), len(df)); bs.append(partial_spearman(df.s.values[i], df.b.values[i], df.z.values[i]))
                pc.append(dict(excl=exn, domain=d, indicator=ind, n=len(df), raw_rho=float(np.corrcoef(rankdata(df.s), rankdata(df.b))[0, 1]), partial_rho=r, lo=np.nanpercentile(bs, 2.5), hi=np.nanpercentile(bs, 97.5)))
    PC = pd.DataFrame(pc); PC.to_csv(S8/f"country_partial_spearman{sfx}.csv", index=False, encoding="utf-8-sig")
    print("\n[국가 수준 편 순위상관(전 기간 풀링, 제외 없음)]\n", PC[PC.excl == "none"].round(3).to_string(index=False))
    print("\n[제외 민감도: 편 순위상관 부호 유지]\n", PC.pivot_table(index=["domain", "indicator"], columns="excl", values="partial_rho").round(3).to_string())
    print("※ 국가 단위 상관은 생태학적 오류 가능, 인과·필요 충족 증거 아님. 부트스트랩은 국가 단위라 논문 단위 불확실성은 미반영. 두 지표 방향이 일관하고 구간이 0을 배제하는 도메인만 '정렬 단서'로 서술.")

    outs = sorted(f.name for f in S8.glob("*.csv")); (S8/f"S8_meta{sfx}.json").write_text(json.dumps(dict(spec_sha256=spec_hash(SPEC), at=now(), quick=QUICK, outputs={f: sha_f(S8/f) for f in outs}), indent=1), encoding="utf-8")
    if not QUICK:
        with (S8/"CHANGELOG_S8.md").open("a", encoding="utf-8") as f:
            f.write(f"- {now()} | S8 align 실행(v2 규칙 해시 {spec_hash(SPEC)[:12]}, 판정 변경 없음): 통과 도메인 {dom_ok}, 렌즈 2종, 제외 민감도 3종(사전 명시), 지표 일관·편 순위상관 보고. 차량은 가용 연도 값, 교통 CO2는 민간항공·수운 포함 | S8 | 본문\n")
    print("저장 완료")

try:
    for s in STEPS: {"freeze": freeze, "tag": tag, "burden": burden, "sample": sample, "kappa": kappa_step, "align": align}[s]()
except Halt as e: print(f"⛔ 중단: {e}")
````
